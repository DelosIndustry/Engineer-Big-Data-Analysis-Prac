#### 판다스와 데이터프레임  
빅분기 실기에서는 판다스는 거의 모든 풀이의 출발점이다. 시험에서 CSV 파일이 주어지면 먼저 데이터를 읽고,  
그 데이터를 표 형태로 확인한 뒤 필요한 계산을 진행한다. 이때 CSV 파일을 읽어 파이썬에서 다루기 쉬운 표로 만들어 주는 도구가 판다스이다.  

데이터프레임은 파이썬에서 코드로 다루는 표라고 이해하면 된다!

```python
import pandas as pd

df = pd.read_csv('./data/basic_students.csv')
```

```python
# 데이터 확인 방법
print(type(df))
print(df.shape)
print(df.head())
```

#### 판다스를 불러오는 순서  
판다스를 처음 사용할 때 가장 많이 발생하는 실수는 라이브러리를 불러오기 전에 pd를 쓰는 것이다.

#### 시리즈와 데이터프레임  
판다스로 데이터를 다루다 보면 데이터프레임과 함께 시리즈라는 자료구조를 자주 접하게 되는데 데이터프레임이  
표 전체라면, 시리즈는 표에서 행 하나 또는 열 하나를 꺼냈을 때 나타나는 1차원 자료구조임. 처음에는 데이터프레임과 시리즈를 따로 외우려고 하기보다, 데이터프레임에서 일부를 선택했을 때 결과가 어떤 형태로 바뀌는지 확인하는 편이 낫다. 이 항목에서는 학생 데이터를 활용해 시링즈와 데이터프레임의 차이를 코드 실행 결과로 살펴보자!

```python
score = df['score_math']
print(type(score))
print(score.head(3))
```

```python
# 데이터프레임과의 차이
score = df[['score_math']]
print(type(score))
print(score.head(3))
```

실제 시험에서 한 칼럼만 계산할 때는 시리즈가 편하고, 여러 칼럼을 묶어서 다시 데이터프레임처럼 다룰 때는  
데이터프레임 형태를 유지하는 편이 좋다.

```python
# 여러 컬럼을 선택하는 경우 컬럼명을 리스트로 묶어야 합니다. 여러 컬럼은 2차원 표 형태이므로 결고도 DF임
name_score = df[['name', 'score_math']]

print(type(name_score))
print(name_score.head(3))
```

```python
# 행 선택 결과
# 시리즈는 컬럼을 선택할 때만 나오는 것이 아니라, 데이터프레임에서 한 행을 선택해도 시리즈가 만들어질 수 있다.
first_row = df.loc[0]

print(type(first_row))
print(first_row)
```

#### to_csv() 기본 사용  
데이터프레임을 csv 파일 형태로 저장할 때는 아래 형식을 기본적으로 사용한다.  
데이터프레임.to_csv('저장할_파일명.csv', index=False)
이때 index=False라는 것은 가장 자주 함께 사용하는 옵션으로 데이터프레임의 인덱스를 CSV 파일에 저장하지 않겠다는 의미이다.  
실기에서는 제출 파일을 만들 때는 보통 문제에서 요구한 칼럼만 저장해야 함. 인덱스까지 저장하면 불필요한 열이 추가될 수 있으므로, 특별한 지시가 없다면 index=False를 붙이는 숩관을 들여야 함!

```python
save_df = df[['student_id', 'class_name', 'score_math', 'score_eng']].head(5)

save_df.to_csv('practice_students.csv', index=False)

check = pd.read_csv('practice_students.csv')
print(check)
print(check.shape)
```

한글이 깨질때의 저장 옵션  
파이썬에서 저장한 CSV 파일을 파이썬으로 다시 읽을 때는 기본 설정으로 충분한 경우가 많다. 다만 CSV 파일을  
엑셀에서 열었을 때 한글이 깨지는 경우 encoding='utf-8-sig' 옵션을 사용할 수 있다. 이 옵션은 시험 풀이의  
필수 문법이라기 보다는 파일을 외부 프로그램에서 확인할 때 도움이 되는 보조 옵션이다.
