#### loc와 iloc  
loc와 iloc는 둘 다 데이터프레임에서 행과 열을 선택하는 도구이다. 차이는 선택 기준이다.  
loc는 인덱스 이름과 컬럼명 기준으로 선택하고, iloc은 행과 열의 위치 번호를 기준으로 선택한다.  
loc는 이름표를 보고 찾는 방식에 가깝다. '0번 인덱스의 region 컬럼'처럼 인덱스 이름과 컬럼명을 직접 적는다.  
iloc는 자리번호를 보고 찾는 방식이다. '첫 번째 행, 세 번째 열'처럼 몇 번째 위치인지를 숫자로 적는다.  
둘 다 숫자가 들어갈 수 있어서 헷갈리지만, loc의 숫자는 인덱스 이름이고, iloc의 숫자는 위치번호라고 생각!

```python
print(df.loc[0, 'region'])
```

```python
print(df.loc[0:2, ['order_id', 'region', 'sales_amount']]) # loc에서는 인덱싱할때, 선택한 범위의 끝값도 포함된다!
```

```python
print(df.iloc[0,2]) # 0,2에 위치한 값을 가져옴!
```

```python
print(df.iloc[0:3, [0,2,9]])
```

```python
print(df.loc[5, ['order_id','region', 'sales_amount']]) # loc은 컬럼명을 사용
print(df.iloc[5, [0, 2, 9]]) # iloc은 열 번호를 사용
```

```python
print(df.loc[0:4, 'sales_amount'].sum())
print(df.iloc[0:5, 9].sum()) # iloc에서는 인덱싱 시에 끝 번호 포함 X
```

#### loc과 iloc에서 주의할 점  
1. 쉼표의 의미  
loc와 iloc의 대괄호 안에서 쉼표 왼쪽은 행 선택, 쉼표 오른쪽은 열 선택을 의미한다. df.loc[0:2, ['order_id', 'region']]처럼 작성하면 0부터 2까지의 행에서 두 컬럼만 선택한다.  
2. 컬럼명과 위치 번호  
loc에서는 컬럼며을 문자열로 적는다. iloc에서는 컬럼명을 적지 않고 위치번호를 적는다. df.iloc[:, 'region']이런 식으로 작성하면 안된다!!  
3. 끝값 포함 여부  
loc의 범위 선택은 끝값을 포함하고, iloc의 범위 선택은 끝값을 포함하지 않는다.  

컬럼명이 명확할 때는 loc를 사용하는 편이 안전함!

#### 조건식의 구조  
1. 비교 연산자  
조건식은 칼럼과 값을 비교해 만든다. 예를 들어, region 칼럼이 수도권인지 확인하려면, df['region']=='수도권' 같은 형태로 작성하면 된다! 이때 등호는 하나가 아니라 2개를 사용한다. 하나의 등호는 값을 저장할 때 사용하고, 두 개의 등호는 같은지 비교할 때 사용한다.

```python
cond = df['region']=='수도권'
print(cond.head(8))
```

```python
# 행 선택  
cond = df['region']=='수도권'
print(df[cond].head())
```

바로 위에서 볼 수 있듯이 필터링 이후에도 원래 행 번호가 유지된다. 결과의 첫 행 번호가 5로 보이는 것은 원본 데이터에서 5번 행이 첫 번째로 조건을 만족했기 때문이다.

```python
cond = df['sales_amount'] >= 300000
print(cond.sum()) # 위 조건문이 True인 값의 개수의 합
```

```python
cond = (df['region'] == '수도권') & (df['sales_amount'] >= 300000) # 두 조건을 동시에 만족하는 거 찾기
print(cond.sum())
print(df.loc[cond, ['order_id', 'region', 'category', 'sales_amount']].head())
```

```python
cond = (df['channel']=='web') | (df['category']=='book')
print(cond.sum())
print(df.loc[cond,['order_id', 'channel', 'category', 'qty']].head()) 
```

```python
cond = ~(df['region'] == '수도권')
print(cond.sum())
```
