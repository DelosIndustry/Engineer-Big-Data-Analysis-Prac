#### 1. 실기 시험의 구성  
결국 핵심은 문제가 요구하는 데이터 분석 과정을 파이썬 코드로 수행하는 것이다.  
제공된 데이터 파일을 불러오고 필요한 처리와 분석을 수행한 뒤, 계산 결과를 확인해 문제에 지정한 방식으로 제출한다.  
실기 문제는 작업형 1, 작업형2, 작업형 3으로 구분되는데, 모두 데이터를 사용하지만, 풀이 과정과 최종 결과의 형태는 다르다. 문제의 접근 방법은 대략적으로 아래와 같음.  
1. 작업형과 문제의 요구사항을 확인  
2. 제공된 데이터 파일의 경로를 확인  
3. 조건, 계산 대상, 예측 대상 또는 검정 대상을 확인한다.  
4. 최종 결과를 답안란에 입력하는지 파일로 저장하는지 확인한다.  
5. 코드를 단계별로 실행하며 중간 결과를 확인한다.  
6. 최종 결과의 자료형과 형식을 검증한다.  
7. 문제에서 지정한 방식으로 제출한다.  

문제에 여러 조건이 한 문장으로 이어져 있다면 조건, 처리 대상, 계산 방법, 제출 형태로 나누어 읽어야 함.  
처음부터 긴 코드를 작성하기 보다는 파일을 읽고, 데이터 상태를 확인하고, 조건을 적용하는 과정을 순서대로 실행하는 편이 오류를 찾기 더 쉽다.

#### 02. 작업형의 차이  
| 구분 | 주요 작업 | 주로 사용하는 방법 | 최종 결과|
| --- | --- | --- | --- |
|작업형 1| 조건에 맞는 데이터 처리와 계산 | 필터링, 정렬, 결측치 처리, 집계 | 계산한 값 |
|작업형 2| 머신러닝 모델 학습과 예측 | 전처리, 모델 학습, 평가, 예측 | 예측 결과 CSV |
|작업형 3| 통계적 분석과 결과 해석 | 가설검정, 분산분석, 회귀분석 | 통계량과 해석 결과 |  

작업형 1은 주어진 데이터에서 문제의 조건에 맞는 값을 구하는 유형.  
특정 조건을 만족하는 행을 선택하거나, 데이터를 정렬하고, 결측치나 이상치를 처리한 뒤 평균, 합계, 개수, 최댓값과 같은 결과를 계산한다. 하나의 문제가 여러 단계로 이어질 수 있으므로 앞의 단계에서 만든 결과를 다음 단계에서 정확하게 사용해야 함. 최종적으로 문제에서 지정한 값을 자료형을 확인해 답안란에 입력한다.  

작업형 2는 학습 데이터로 ML 모델을 만든 뒤에 평가용 데이터의 결과를 예측하는 유형이다. 먼저 예측 대상인 target 칼럼을 확인하고, 데이터 전처리와 모델 학습을 진행한다. 모델은 학습한 뒤에는 test 데이터의 예측값을 생성하고, 문제에서 요구한 파일명과 칼럼명으로 CSV 파일을 저장한다. 분류와 회귀 중 어떤 문제인지, 클래스값과 확률값 중에 무엇을 제출해야 하는지 반드시 확인해야 함.  

작업형 3는 통계적 방법을 이용해서 데이터를 분석하고 결과를 해석하는 유형이다. 모집단과 표본, 귀무가설고 대립가설, 검정통계량, p-value와 같은 개념을 이해하고 문제에 적합한 통계함수를 사용해야 하며 문제에서 요구한 반올림 자릿수와 결론의 표현도 함께 확인해야 함.
