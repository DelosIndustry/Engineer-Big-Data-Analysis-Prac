## 01 조건 필터링

### 조건식과 불리언 값

#### 1. 데이터 불러오기

실습에 사용할 데이터를 불러와 크기와 앞부분을 확인한다.

```python
import pandas as pd
filter_df = pd.read_csv('data/part02_type1/type1_filter_sort.csv')
print(filter_df.shape)
print(filter_df[['customer_id', 'age', 'region', 'score',
                 'purchase_count', 'total_spend', 'segment']].head().to_string(index=False))
```

#### 2. 조건에 맞는 행 선택

region이 central인 행을 찾는 조건식을 만들어본다. 조건식의 결과가 어떤 형태인지 먼저 확인한다.

```python
cond = filter_df['region'] == 'central'
print(cond.head())
print(cond.sum())
```

#### 3. 특정 컬럼 계산

조건에 맞는 행에서 특정 컬럼만 계산할 때는 loc를 함께 사용하면 계산 대상을 분명하게 지정할 수 있다.

```python
cond = filter_df['score'] >= 85
answer = filter_df.loc[cond, 'total_spend'].sum()
print(cond.sum())
print(answer)
```

## 02 다중 조건 필터링

### 1 다중 조건 필터링이란

#### 1. AND 조건

두 조건을 모두 만족하는 행을 찾을 때는 &로 조건을 연결한다. region이 central이면서 score가 80 이상인 행 수를 세어 본다.

```python
cond = (filter_df['region'] == 'central') & (filter_df['score'] >= 80)
answer = cond.sum()
print(answer)
```

#### 2. OR 조건

둘 중 하나만 만족해도 되는 조건은 |로 연결한다. region이 north이거나 south인 행 수를 세어 본다.

```python
cond = (filter_df['region'] == 'north') | (filter_df['region'] == 'south')
answer = cond.sum()
print(answer)
```

### 2 연산 순서와 여러 값 선택

#### 1. 괄호와 연산 순서

#### 2. 여러 값 선택

같은 컬럼에서 여러 값 중 하나에 해당하는 행을 선택할 때는 | 조건을 반복하기보다 isin( )을 사용한다. filter_df['segment'].isin(['A', 'B'])는 segment 값이 A 또는 B인 행을 True로 표시한다.

```python
cond = (
    filter_df['segment'].isin(['A', 'B'])
    & (filter_df['total_spend'] >= 300000)
)
answer = filter_df.loc[cond, 'purchase_count'].sum()
print(cond.sum())
print(answer)
```

## 05 결측치 처리

### 1 결측치 처리 과정

#### 1. 데이터 불러오기

결측치와 이상치 실습에 사용할 데이터를 불러와 크기와 앞부분을 확인한다.

```python
missing_df = pd.read_csv('data/part02_type1/type1_missing_outlier.csv')
print(missing_df.shape)
print(missing_df[['row_id', 'item_group', 'price', 'qty',
                  'rating', 'views', 'event_flag']].head(10).to_string(index=False))
```

#### 2. 결측치 개수 확인

isna( )와 sum( )을 조합하면 컬럼별 결측치 개수와 전체 결측치 개수를 확인할 수 있다.

```python
print(missing_df.isna().sum())
print(missing_df.isna().sum().sum())
```

#### 3. 결측치 대체

rating의 결측치를 중앙값으로 대체해본다. 원본을 유지하기 위해 복사본을 만든 뒤 fillna( )를 적용한다.

```python
filled = missing_df.copy()
median_rating = filled['rating'].median()
filled['rating'] = filled['rating'].fillna(median_rating)
print(median_rating)
print(round(filled['rating'].mean(), 2))
```

#### 4. 결측치 제거

결측치가 있는 행을 제거할 때는 dropna( )를 사용한다. subset에 기준 컬럼을 지정하면 그 컬럼에 결측치가 있는 행만 제거된다.

```python
complete = missing_df.dropna(subset=['views'])
print(len(complete))
print(complete['qty'].sum())
```

#### 5. 결측치 처리 순서

‘전체 중앙값으로 대체한 뒤 조건을 적용’하는지, ‘특정 조건의 데이터만 고른 뒤 그 안에서 중앙값을 구해 대체’하는지에 따라 값이 달라진다. 조건과 결측치 처리의 순서를 문제 문장에 맞춘다.

```python
filled = missing_df.copy()
filled['views'] = filled['views'].fillna(filled['views'].median())
cond = filled['qty'] >= 10
answer = int(filled.loc[cond, 'views'].mean())
print(answer)
```

### 2 그룹 기준 결측치 대체하기

#### 1. 그룹별 대푯값으로 결측치 대체하기

전체 평균이나 중앙값으로 결측치를 채우면 모든 결측 행에 같은 값이 들어간다. 그룹마다 값의 분포가 다르다면 각 행이 속한 그룹의 대표값을 사용해 결측치를 채울 수 있다.

다음 예제에서는 item_group별 rating 중앙값을 계산한다. groupby( ).median( )은 그룹 수만큼의 결과를 만들지만, transform(‘median’)은 원본과 같은 길이와 인덱스로 각 행에 해당하는 중앙값을 반환한다. 따라서 fillna( )에 바로 연결할 수 있다. 만약 기준이 둘 이상이면 groupby(['컬럼1', '컬럼2'])처럼 리스트로 넘긴다.

```python
group_median = missing_df.groupby(
    'item_group'
)['rating'].transform('median')
filled = missing_df.copy()
filled['rating'] = filled['rating'].fillna(group_median)
print(filled['rating'].isna().sum())
print(
    filled.loc[
        missing_df['rating'].isna(),
        ['row_id', 'item_group', 'rating']
    ].head(3).to_string(index=False)
)
```

#### 2. 두 개 이상 기준으로 결측치 대체하기

기준이 하나로 부족하면 여러 컬럼을 함께 묶을 수 있다. groupby( )에 리스트로 컬럼을 넘기면 모든 기준이 같은 행끼리 하나의 그룹이 된다. 기준이 세분화될수록 각 행에 더 가까운 값이 채워진다.

```python
group_median = missing_df.groupby(
['item_group', 'event_flag']
)['rating'].transform('median')
filled = missing_df.copy()
filled['rating'] = filled['rating'].fillna(group_median)
print(filled['rating'].isna().sum())
print(
filled.loc[
missing_df['rating'].isna(),
['row_id', 'item_group', 'event_flag', 'rating']
].head(3).to_string(index=False)
)
```

## 06 이상치 처리

### 1 이상치 처리

#### 1. IQR 기준

#### 2. IQR 상한 초과값 확인과 제외

IQR 방식으로 상한을 계산하고 상한을 초과하는 값을 제외해본다. 1사분위수와 3사분위수를 구한 뒤 Q3 + 1.5 × IQR을 상한으로 사용한다.

```python
q1 = missing_df['price'].quantile(0.25)
q3 = missing_df['price'].quantile(0.75)
iqr = q3-q1
upper = q3+1.5*iqr
normal = missing_df[missing_df['price'] <= upper]
print(q1, q3, upper)
print(missing_df.shape[0], normal.shape[0])
print(int(normal['price'].mean()))
```

### 2 이상치 대체

#### 1. 상한값으로 이상치 대체

이상치를 제외하는 대신 상한값으로 눌러서 대체하는 방법도 있다. 상한을 초과하는 price를 upper 값으로 바꿔 본다.

```python
capped = missing_df.copy()
capped.loc[capped['price'] > upper, 'price'] = upper
print(capped['price'].max())
print(round(capped['price'].mean(), 2))
```

#### 2. 조건을 함께 적용하기

이상치 처리 결과는 조건 계산으로 이어지는 경우가 많다. 상한 이하인 데이터에서 event_flag가 1인 행의 price 평균을 정수로 구해본다.

```python
normal = missing_df[missing_df['price'] <= upper]
cond = normal['event_flag'] == 1
answer = int(normal.loc[cond, 'price'].mean())
print(answer)
```

## 07 파생변수 생성 후 계산

### 1 파생변수 생성

#### 1. 데이터 불러오기

파생변수와 그룹 집계 실습에 사용할 데이터를 불러와 크기와 앞부분을 확인한다.

```python
sales_df = pd.read_csv('data/part02_type1/type1_derived_group.csv')
print(sales_df.shape)
print(sales_df[['order_id', 'region', 'category', 'qty',
                'unit_price', 'coupon_rate', 'net_sales']].head().to_string(index=False))
```

#### 2. 계산형 파생변수

수량과 단가를 곱해 할인 전 매출액 컬럼을 만들어본다. 기존 컬럼끼리 연산한 결과를 새 컬럼 이름에 저장하면 된다.

```python
sales_df['gross_sales'] = sales_df['qty'] * sales_df['unit_price']
print(sales_df[['order_id', 'qty', 'unit_price', 'gross_sales']].head())
```

#### 3. 정수형 파생변수

계산 결과에 소수가 생기면 문제 지시에 따라 정수로 만들어야 한다. 컬럼 전체를 아래쪽 정수로 내릴 때는 np.floor( )를 적용한 뒤 astype(int)로 정수형으로 바꾼다. 순매출액을 만 원 단위로 바꾸고 소수 이하를 버려본다.

```python
import numpy as np
sales_df['gross_sales_10k'] = (
    np.floor(sales_df['gross_sales'] / 10000).astype(int)
)
print(sales_df[['order_id', 'gross_sales', 'gross_sales_10k']].head().to_string(index=False))
```

### 2 파생변수 계산

#### 1. 할인 금액 계산

할인 전 매출액에 쿠폰 할인율을 곱하면 할인 금액이 된다. 이어서 할인 금액을 빼 순매출액도 계산해본다.

```python
sales_df['discount_amount'] = (
    sales_df['gross_sales'] * sales_df['coupon_rate']
)
sales_df['calc_net_sales'] = (
    sales_df['gross_sales'] - sales_df['discount_amount']
)
print(int(sales_df['discount_amount'].sum()))
```

#### 2. 조건형 파생변수

조건식의 결과를 파생변수로 저장할 수도 있다. 할인 전 매출액이 500000 이상이면 1, 아니면 0이 되는 컬럼을 만들어본다.

```python
sales_df['high_order'] = (sales_df['gross_sales'] >= 500000).astype(int)
print(sales_df['high_order'].sum())
print(sales_df.groupby('category')['high_order'].sum().sort_values(ascending=False))
```

#### 3. 파생변수를 이용한 계산

만들어둔 파생변수는 이후 계산에 바로 사용할 수 있다. metro 지역의 할인 전 매출액 합계를 구해본다.

```python
cond = sales_df['region'] == 'metro'
answer = sales_df.loc[cond, 'gross_sales'].sum()
print(answer)
```

#### 4. 수치 범위를 구간으로 나누기

수량이나 점수처럼 연속된 숫자를 일정한 범위로 나누면 범주형 파생변수를 만들 수 있다. pd.cut( )의 bins에는 구간 경계를, labels에는 각 구간에 붙일 이름을 지정한다.

right=False를 지정하면 왼쪽 경계는 포함하고 오른쪽 경계는 제외한다. 마지막 경계에 np.inf를 사용하면 데이터의 최댓값이 커져도 마지막 구간에서 빠지지 않는다.

```python
bins = [0, 4, 8, 12, np.inf]
labels = ['3개이하', '4~7개', '8~11개', '12개이상']
sales_df['qty_group'] = pd.cut(
    sales_df['qty'],
    bins=bins,
    labels=labels,
    right=False
)
qty_summary = (
    sales_df.groupby('qty_group', observed=False)['net_sales']
    .agg(['count', 'mean'])
    .round(1)
)
print(qty_summary)
print(sales_df['qty_group'].isna().sum())
```

#### 5. 계획값과 실젯값의 차이 비교하기

계획값과 실젯값의 차이는 실젯값이 더 크면 양수, 더 작으면 음수가 된다. 어느 방향으로 벗어났는지보다 차이의 크기가 중요하면 abs( )로 절댓값을 구한 뒤 idxmax( )를 사용한다.

```python
gap_df = pd.DataFrame({
    'branch': ['A지점', 'B지점', 'C지점', 'D지점'],
    'planned_orders': [320, 280, 350, 300],
    'actual_orders': [305, 310, 330, 260]
})
gap_df['gap'] = (
    gap_df['actual_orders'] - gap_df['planned_orders']
)
idx = gap_df['gap'].abs().idxmax()
answer = gap_df.loc[idx, 'branch']
print(gap_df.to_string(index=False))
print(answer)
```

## 08 그룹별 집계

### 1 그룹별 집계란

#### 1. 하나의 기준으로 집계

지역별 순매출액 합계부터 계산해본다. 합계가 가장 큰 지역과 그 값은 idxmax( )와 max( )로 확인한다.

```python
region_sum = sales_df.groupby('region')['net_sales'].sum()
print(region_sum)
print(region_sum.idxmax())
print(region_sum.max())
```

#### 2. 그룹별 최댓값 행 다시 선택하기

집계 결과에서 가장 큰 그룹명을 찾는 것과 각 그룹에서 가장 큰 값을 가진 원본 행을 찾는 것은 다르다. 각 region에서 net_sales가 가장 큰 주문의 order_id나 category를 확인하려면 원본 행의 인덱스가 필요하다.

groupby( )[계산 컬럼].idxmax( )는 각 그룹에서 최댓값을 가진 원본 행의 인덱스를 반환한다. 이 인덱스 목록을 sales_df.loc[]에 넣으면 해당 주문의 다른 컬럼도 함께 선택할 수 있다. 그룹 안에 최댓값이 여러 개라면 idxmax( )는 먼저 나타난 행의 인덱스를 반환한다.

```python
idx = sales_df.groupby('region')['net_sales'].idxmax()
top_rows = sales_df.loc[idx].sort_values('region')
print(
    top_rows[
        ['region', 'order_id', 'category', 'net_sales']
    ].to_string(index=False)
)
answer = (top_rows['category'] == 'stationery').sum()
print(answer)
```

#### 3. 집계 결과 정렬

집계 결과에 sort_values( )를 이어 붙이면 큰 값부터 정렬된 결과를 얻을 수 있다.

```python
category_sum = (
    sales_df.groupby('category')['net_sales']
   .sum()
   .sort_values(ascending=False)
)
print(category_sum)
```

### 2 두 개 기준으로 집계

#### 1. 두 가지의 기준으로 집계

그룹 기준을 리스트로 전달하면 지역과 카테고리 조합별로 집계할 수 있다.

```python
qty_sum = (
    sales_df.groupby(['region', 'category'])['qty']
   .sum()
   .sort_values(ascending=False)
)
print(qty_sum.head())
```

#### 2. 두 기준의 값을 표로 펼쳐 비교하기

region과 category처럼 두 기준으로 나눈 값을 같은 행에서 비교하려면 한 기준을 행, 다른 기준을 열에 배치한 wide 형태가 편리하다. pivot_table( )은 원본 데이터에서 이러한 표를 바로 만들 수 있다.

다음 예제는 region을 행, category를 열로 두고 net_sales 평균을 계산한다. 각 region에서 평균이 가장 큰 category를 찾으려면 같은 행의 여러 열을 비교해야 하므로 idxmax(axis=1)을 사용한다.

```python
wide = sales_df.pivot_table(
    index='region',
    columns='category',
    values='net_sales',
    aggfunc='mean'
)
best_category = wide.idxmax(axis=1)
print(best_category)
```

이미 groupby([‘region’, ‘category’])로 만든 다중 인덱스 결과가 있다면 unstack( )으로 안쪽 인덱스를 열로 펼칠 수도 있다. pivot_table( )은 원본 데이터에서 바로 wide 표를 만들 때 사용하고, unstack( )은 만들어진 다중 인덱스 결과의 형태를 바꿀 때 사용한다.

```python
region_category_mean = (
    sales_df.groupby(['region', 'category'])['net_sales']
    .mean()
    .unstack()
)
```

#### 3. 여러 통계량 확인

합계, 평균, 개수를 한 표에서 비교하고 싶다면 agg( )에 함수 목록을 전달한다.

```python
region_summary = (
    sales_df.groupby('region')['net_sales']
   .agg(['count', 'sum', 'mean'])
   .round(1)
)
print(region_summary)
```

## 09 날짜 데이터와 문자열 처리

### 1 날짜 데이터 처리

#### 1. 데이터 불러오기

날짜와 문자열 실습에 사용할 데이터를 불러와 크기와 앞부분을 확인한다.

```python
date_df = pd.read_csv('data/part02_type1/type1_date_string.csv')
print(date_df.shape)
print(date_df[['reservation_id', 'service_code', 'reserved_at',
               'cancelled_at', 'status']].head().to_string(index=False))
```

#### 2. 날짜형 변환

reserved_at과 cancelled_at은 문자열 상태이므로 pd.to_datetime( )으로 날짜형으로 변환한다.

```python
date_df['reserved_at'] = pd.to_datetime(date_df['reserved_at'])
date_df['cancelled_at'] = pd.to_datetime(date_df['cancelled_at'])
print(date_df[['reserved_at', 'cancelled_at']].dtypes)
```

#### 3. 날짜 요소 추출

날짜형 컬럼에서는 dt 접근자로 월, 요일, 시간을 각각 꺼낼 수 있다.

```python
date_df['reserved_month'] = date_df['reserved_at'].dt.month
date_df['reserved_dayofweek'] = date_df['reserved_at'].dt.dayofweek
date_df['reserved_hour'] = date_df['reserved_at'].dt.hour
print(date_df[['reserved_at', 'reserved_month',
               'reserved_dayofweek', 'reserved_hour']].head())
```

#### 4. 주말 조건

dt.dayofweek는 월요일을 0, 일요일을 6으로 나타낸다. 따라서 값이 5 이상인 행을 선택하면 토요일과 일요일에 해당하는 예약만 고를 수 있다. 시험에서는 요일 이름으로 변환하기보다 숫자로 비교하는 편이 코드가 단순하다.

```python
weekend = date_df['reserved_dayofweek'] >= 5
print(weekend.sum())
```

#### 5. 날짜 차이

두 날짜형 컬럼의 차이를 계산해본다. 예약일에서 취소일을 뺀 뒤 dt.days로 일 단위 차이를 꺼낸다.

```python
date_df['reserved_date'] = date_df['reserved_at'].dt.normalize()
has_cancel = date_df['cancelled_at'].notna()
date_df.loc[has_cancel, 'cancel_gap'] = (
    date_df.loc[has_cancel, 'reserved_date']
    - date_df.loc[has_cancel, 'cancelled_at']
).dt.days
print(round(date_df.loc[has_cancel, 'cancel_gap'].mean(), 2))
```

#### 6. 시간대 조건

추출해둔 시간 컬럼은 조건식에 바로 사용할 수 있다. 20시에 예약된 건수를 세어 본다.

```python
cond = date_df['reserved_hour'] == 20
print(cond.sum())
```
