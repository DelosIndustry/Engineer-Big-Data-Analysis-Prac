## 09 조건 필터링

### 1 조건식의 구조

#### 1. 비교 연산자

조건식은 컬럼과 값을 비교해 만든다. 예를 들어 region 컬럼이 수도권인지 확인하려면

df[‘region’] == ‘수도권’형태로 작성한다. 이때 등호는 하나가 아니라 두 개를 사용한다. 하나의 등호는 값을 저장할 때 사용하고, 두 개의 등호는 같은지 비교할 때 사용한다.

````python
cond = df['region'] == '수도권'
print(cond.head(8))
````

#### 2. 행 선택

조건식을 데이터프레임 안에 넣으면 True인 행만 선택된다. 다음 코드는 region이 수도권인 주문만 남긴다.

````python
cond = df['region'] == '수도권'
print(df[cond].head())
````

#### 3. 개수 확인

조건에 맞는 행이 몇 개인지만 확인할 때는 조건식에 sum( )을 적용할 수 있다. True는 1, False는 0처럼 계산되므로 조건을 만족하는 행 수를 빠르게 확인할 수 있다.

````python
cond = df['sales_amount'] >= 300000
print(cond.sum())
````

### 2 여러 조건 결합

#### 1. AND 조건

두 조건을 모두 만족해야 할 때는 & 기호를 사용한다. 각 조건은 반드시 소괄호로 묶는다. 다음 코드는 region이 수도권이면서 sales_amount가 300000 이상인 주문을 고른다.

````python
cond = (df['region'] == '수도권') & (df['sales_amount'] >= 300000)
print(cond.sum())
print(df.loc[cond, ['order_id', 'region', 'category', 'sales_amount']].head())
````

#### 2. OR 조건

둘 중 하나만 만족해도 되는 조건은 | 기호를 사용한다. 다음 코드는 channel이 web이거나 category가 book인 주문을 고른다.

````python
cond = (df['channel'] == 'web') | (df['category'] == 'book')
print(cond.sum())
print(df.loc[cond, ['order_id', 'channel', 'category', 'qty']].head())
````

#### 3. NOT 조건

특정 조건을 만족하지 않는 행을 고를 때는 조건식 앞에 ~ 기호를 붙인다. 다음 코드는 region이 수도권이 아닌 주문을 고른다.

````python
cond = ~(df['region'] == '수도권')
print(cond.sum())
````

### 3 조건식 작성 시 주의할 점

#### 1. 등호 사용

조건을 비교할 때는 ==를 사용한다. 등호를 하나만 쓰면 비교가 아니라 값을 넣는 의미가 되므로 원하는 필터링 코드가 아니다.

> 설명용 코드입니다. 그대로 실행되지 않습니다.

```python
df['region'] = '수도권'   # 잘못된 코드: region 전체가 '수도권'으로 바뀐다
df['region'] == '수도권'  # 수정한 코드
```

#### 2. 괄호 사용

&와 |를 사용할 때는 각 조건을 소괄호로 묶는다. 소괄호를 생략하면 판다스가 조건식을 의도와 다르게 해석하거나 오류가 발생할 수 있다.

> 설명용 코드입니다. 그대로 실행되지 않습니다.

```python
df[df['region'] == '수도권' & df['qty'] >= 5]    # 잘못된 코드
df[(df['region'] == '수도권') & (df['qty'] >= 5)]  # 수정한 코드
```

#### 3. 문자열 따옴표

문자열 값을 비교할 때는 따옴표가 필요하다. 데이터에 들어 있는 문자값을 정확히 적는다.

> 설명용 코드입니다. 그대로 실행되지 않습니다.

```python
df[df['region'] == 수도권] # 잘못된 코드: NameError 발생
df[df['region'] == '수도권']  # 수정한 코드
```

## 10 정렬

### 1 정렬 기준

#### 1. 오름차순

sort_values( )는 지정한 컬럼을 기준으로 데이터를 정렬한다. 기본값은 오름차순이다. 숫자라면 작은 값에서 큰 값 순서로 정렬된다. 이 데이터처럼 날짜가 YYYY-MM-DD 형식의 문자열이면 빠른 날짜에서 늦은 날짜 순서로 정렬된다.

````python
result = df.sort_values('sales_amount')
print(result[['order_id', 'region', 'category', 'sales_amount']].head())
````

#### 2. 내림차순

큰 값부터 보고 싶다면 ascending=False를 사용한다. 작업형 1에서 최댓값에 해당하는 행을 찾을 때 자주 쓰는 형태이다.

````python
result = df.sort_values('sales_amount', ascending=False)
print(result[['order_id', 'region', 'category', 'sales_amount']].head())
````

### 2 여러 기준 정렬

#### 1. 기준 컬럼 두 개

정렬 기준이 두 개 이상이면 리스트로 컬럼명을 전달한다. 다음 코드는 region을 먼저 오름차순으로 정렬하고, 같은 region 안에서는 sales_amount가 큰 순서로 정렬한다.

````python
result = df.sort_values(['region', 'sales_amount'], ascending=[True, False])
print(result[['order_id', 'region', 'category', 'sales_amount']].head(8))
````

#### 2. 인덱스 번호

정렬을 해도 원래 인덱스 번호는 그대로 남는다. 그래서 내림차순 정렬 결과의 첫 행 인덱스가 0이 아니라 59로 보일 수 있다. 정렬 후 새 번호를 붙여야 한다면 reset_index(drop=True)를 함께 사용한다.

````python
result = df.sort_values('sales_amount', ascending=False).reset_index(drop=True)
print(result[['order_id', 'region', 'category', 'sales_amount']].head())
````

### 3 정렬 후 값 추출

#### 1. 첫 번째 행

정렬은 행을 다시 배치하는 작업이므로, 정렬 후 iloc[0]을 사용하면 정렬 기준에서 첫 번째에 해당하는 행을 가져올 수 있다.

````python
result = df.sort_values('sales_amount', ascending=False)
first = result.iloc[0]
print(first[['order_id', 'sales_amount']])
````

#### 2. 조건과 정렬

조건 필터링과 정렬은 함께 자주 사용된다. 다음 코드는 category가 book인 주문만 남긴 뒤 sales_amount가 큰 순서로 정렬한다.

````python
book = df[df['category'] == 'book']
result = book.sort_values('sales_amount', ascending=False)
print(result[['order_id', 'category', 'sales_amount']].head(5))
````
