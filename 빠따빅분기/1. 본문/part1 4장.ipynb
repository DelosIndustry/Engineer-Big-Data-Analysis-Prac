### 3 컬럼별 결측치

#### 1. isna( )

isna( )는 값이 비어 있는지 확인한다. 값이 비어 있으면 True, 값이 있으면 False를 반환한다. 데이터프레임 전체에 적용하면 같은 모양의 True/False 결과가 만들어진다.

#### 2. isna( ).sum( )

컬럼별 결측치 개수는 isna( ).sum( )으로 확인한다. True는 1, False는 0처럼 계산되므로 컬럼마다 True의 개수가 합산된다.

```python
print(df.isna().sum())
```

### 2 지출액 이상치 처리

#### 1. 기준 계산

먼저 spend 컬럼의 IQR 기준을 계산한다. 1사분위수와 3사분위수를 구하고, 그 차이인 IQR을 이용해 하한값과 상한값을 만든다.

```python
q1 = df['spend'].quantile(0.25)
q3 = df['spend'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
print(q1, q3, lower, upper)
```

#### 2. 행 제거

문제에서 이상치를 제거하라고 하면 조건에 맞는 행을 제외한다. 이때 spend가 결측치인 행까지 함께 사라지지 않도록 조건을 조심해서 작성한다.

```python
cond = df['spend'] > upper
print(df.loc[cond, cols])
df_removed = df[(df['spend'].isna()) | (df['spend'] <= upper)]
print(df_removed.shape)
print(df_removed['spend'].max())
```

#### 3. 상한값 대체

행을 제거하지 않고 값을 제한해야 하는 문제도 있다. 이때는 원본을 복사한 뒤, 상한값을 넘는 값만 상한값으로 바꾼다.

```python
df_capped = df.copy()
df_capped.loc[df_capped['spend'] > upper, 'spend'] = upper
print(df_capped.loc[df_capped['member_id'] == 'MB0002', cols])
print(df_capped['spend'].max())
```

### 3 나이 이상치 처리

age 컬럼도 같은 방식으로 IQR 기준을 계산할 수 있다. 나이처럼 의미가 분명한 컬럼은 무조건 삭제하기보다 문제에서 요구한 방식에 맞추어 처리한다.

```python
q1 = df['age'].quantile(0.25)
q3 = df['age'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
cond = (df['age'] < lower) | (df['age'] > upper)
print(lower, upper)
print(df.loc[cond, cols])
```

#### 2. 중앙값 대체

문제에서 이상치를 중앙값으로 대체하라고 하면 먼저 중앙값을 구한 뒤, 이상치 조건에 해당하는 값만 바꾼다.

```python
median_age = df['age'].median()
df_age_fixed = df.copy()
df_age_fixed.loc[cond, 'age'] = median_age
print(median_age)
print(df_age_fixed.loc[cond, cols])
```

## 06 파생변수 생성

### 1 원본 컬럼 확인

#### 1. 계산에 사용할 컬럼

이번 실습에서는 주문 데이터인 data/part01_pandas/pandas_orders.csv 파일을 사용한다. 수량, 단가, 할인율, 주문일, 매출액이 함께 들어 있어 파생변수 생성 흐름을 확인하기에 적합하다.

```python
orders = pd.read_csv('data/part01_pandas/pandas_orders.csv')
cols = ['order_id', 'qty', 'unit_price', 'discount_rate', 'order_date', 'sales_amount']
print(orders[cols].head())
```

### 2 계산 컬럼 생성

#### 1. 수량과 단가의 곱

가장 기본적인 파생변수는 컬럼끼리 계산해서 만든다. 수량과 단가를 곱하면 할인 전 주문 금액을 만들 수 있다.

```python
df_calc = orders.copy()
df_calc['gross_amount'] = df_calc['qty'] * df_calc['unit_price']
print(df_calc[['order_id', 'qty', 'unit_price', 'gross_amount', 'sales_amount']].head())
```

#### 2. 할인액과 결제 금액

할인율이 있는 경우에는 할인 전 금액에 할인율을 곱해 할인액을 만들 수 있다. 이후 할인 전 금액에서 할인액을 빼면 결제 금액을 계산할 수 있다.

```python
df_calc['discount_amount'] = (df_calc['gross_amount'] * df_calc['discount_rate']).round(
).astype(int)
df_calc['pay_amount'] = df_calc['gross_amount'] - df_calc['discount_amount']
print(df_calc[['order_id', 'gross_amount', 'discount_rate', 'discount_amount', 'pay_amount',
'sales_amount']].head())
```

### 3 날짜 파생변수 생성

#### 1. 주문 월 추출

날짜 컬럼에서 연도, 월, 일을 꺼내 새 컬럼으로 만들 수도 있다. order_date는 CSV를 읽은 직후에는 문자열이므로, 먼저 날짜형으로 바꾼 뒤 월을 추출한다.

```python
df_date = orders.copy()
df_date['order_date'] = pd.to_datetime(df_date['order_date'])
df_date['order_month'] = df_date['order_date'].dt.month
print(df_date[['order_id', 'order_date', 'order_month']].head())
print(df_date['order_month'].value_counts().sort_index())
```

### 4 조건 결과 저장

#### 1. 큰 주문 표시

조건식의 결과도 새 컬럼으로 저장할 수 있다. 예를 들어 매출액이 500000원 이상인 주문을 큰 주문으로 표시하면, 이후 조건 필터링이나 집계에 다시 사용할 수 있다.

```python
df_flag = orders.copy()
df_flag['large_order'] = df_flag['sales_amount'] >= 500000
print(df_flag[['order_id', 'sales_amount', 'large_order']].head(8))
print(df_flag['large_order'].value_counts())
```

replace( )는 특정 값을 다른 값으로 바꿀 때 사용한다. 성별 컬럼에서 f는 F로, m은 M으로 바꾸고, unknown은 U로 바꾸어 표기를 통일한다.

```python
df_fix = df.copy()
df_fix['gender'] = df_fix['gender'].replace({'f': 'F', 'm': 'M', 'unknown': 'U'})
print(df_fix['gender'].value_counts(dropna=False))
```

#### 2. 문자열 메서드를 이용한 대문자 통일

문자열 값 전체를 같은 규칙으로 바꿀 때는 str.upper( )나 str.lower( )를 사용할 수 있다. 등급 컬럼은 일부 값만 소문자로 되어 있으므로 전체를 대문자로 통일한다.

```python
df_fix['grade'] = df_fix['grade'].str.upper()
print(df_fix['grade'].value_counts(dropna=False))
```

### 3 조건에 맞는 값 수정

#### 1. loc를 이용한 선택 변경

조건에 맞는 행만 골라 특정 컬럼의 값을 바꿀 때는 loc를 사용한다. 예를 들어 이용 금액이 500000원 이상인 회원의 등급을 VIP로 바꿀 수 있다.

```python
cond = df_fix['spend'] >= 500000
print(df_fix.loc[cond, ['member_id', 'spend', 'grade']].head())
df_fix.loc[cond, 'grade'] = 'VIP'
print(df_fix.loc[cond, ['member_id', 'spend', 'grade']].head())
print(df_fix['grade'].value_counts())
```

### 4 비어 있는 표싯값 입력

#### 1. 메모 값 보완

결측치를 분석에서 제외할 수도 있지만, 문자 컬럼에서는 빈 값을 하나의 표싯값으로 바꿔 관리하기도 한다. memo 컬럼의 빈 값을 기록없음으로 바꾸면 이후 value_counts( )에서 별도 항목으로 확인할 수 있다.

```python
print(df_fix['memo'].isna().sum())
df_fix.loc[df_fix['memo'].isna(), 'memo'] = '기록없음'
print(df_fix[['member_id', 'memo']].head(8))
print(df_fix['memo'].isna().sum())
```

### 5 매핑 테이블 활용

#### 1. 등급명 변환

값마다 대응되는 설명을 붙이고 싶을 때는 map( )을 사용할 수 있다. 등급 코드를 한글 설명으로 바꿔 새 컬럼에 저장하면 결과를 읽기 쉬워진다.

```python
grade_map = {'BASIC': '일반', 'SILVER': '우수', 'GOLD': '골드', 'VIP': '최우수'}
df_fix['grade_label'] = df_fix['grade'].map(grade_map)
print(df_fix[['member_id', 'grade', 'grade_label']].head(10))
print(df_fix['grade_label'].value_counts(dropna=False))
```

#### 1. 중복 묶음 출력

중복 행의 개수만 확인하면 어떤 데이터가 반복되었는지 알 수 없다. keep=False를 사용해 중복 묶음 전체를 출력하면 어떤 회원 정보가 반복되었는지 확인할 수 있다.

### 3 중복 행 제거

#### 1. 기본 제거 방식

drop_duplicates( )는 중복 행을 제거할 때 사용한다. 기본 설정에서는 처음 나온 행은 남기고, 뒤에 나온 중복 행만 제거한다.

```python
df_drop = df.drop_duplicates()
print(df_drop.shape)
print(df_drop.duplicated().sum())
```

중복 행을 제거해도 기존 인덱스 번호는 그대로 남을 수 있다. 행 번호를 0부터 다시 정리하고 싶다면 reset_index(drop=True)를 이어서 사용한다.

```python
df_clean = df.drop_duplicates().reset_index(drop=True)
print(df_clean.shape)
print(df_clean.tail(3))
```

모든 컬럼이 같을 때만 중복으로 볼 수도 있지만, 특정 컬럼을 기준으로 중복을 판단해야 하는 경우도 있다. 예를 들어 회원 ID는 한 회원을 구분하는 값이므로 member_id가 반복되면 같은 회원이 중복 입력되었는지 확인해야 한다.

```python
print(df.duplicated(subset=['member_id']).sum())
cond = df.duplicated(subset=['member_id'], keep=False)
print(df.loc[cond, ['member_id', 'age', 'spend', 'grade']])
```

### 5 남길 행 선택

#### 1. keep 옵션

duplicated( )와 drop_duplicates( )에는 keep 옵션을 사용할 수 있다. keep=‘first’는 첫 번째 행을 남기고 뒤쪽 행을 중복으로 본다. keep=‘last’는 마지막 행을 남기고 앞쪽 행을 중복으로 본다. keep=False는 중복 묶음 전체를 중복으로 표시한다.

```python
print(df.duplicated(subset=['member_id'], keep='first').sum())
print(df.duplicated(subset=['member_id'], keep='last').sum())
print(df.duplicated(subset=['member_id'], keep=False).sum())
```

판다스에서 컬럼 안의 문자열을 처리할 때는 컬럼 뒤에 .str을 붙인다. 파이썬 문자열 하나를 처리하는 것이 아니라, 데이터프레임의 한 컬럼에 들어 있는 여러 문자열을 한꺼번에 처리하기 때문이다.

```python
print(text_df['service_code'].str[:2].head().to_string(index=False))
```

### 3 문자 제거와 분리

#### 1. 전화번호 정리

전화번호처럼 일정한 기호가 들어 있는 문자열은 replace( )나 split( )으로 정리할 수 있다. 하이픈을 제거하면 숫자만 이어진 형태가 되고, 하이픈을 기준으로 나누면 가운데 번호만 따로 가져올 수도 있다.

```python
text_df['phone_clean'] = text_df['phone'].str.replace('-', '', regex=False)
text_df['phone_mid'] = text_df['phone'].str.split('-').str[1]
print(text_df[['phone', 'phone_clean', 'phone_mid']].head(8).to_string(index=False))
```

```python
text_df['service_group'] = text_df['service_code'].str[:2]
text_df['service_no'] = text_df['service_code'].str[-3:].astype(int)
print(text_df[['service_code', 'service_group', 'service_no']].head(8).to_string(index=False))
print(text_df['service_group'].value_counts().sort_index())
```

### 5 문자열 조건

#### 1. 포함 여부 확인

문자열 조건은 특정 글자를 포함하는 행만 골라낼 때 사용한다. 이메일 주소에서 특정 도메인을 포함하는 행을 찾거나, 서비스 코드가 특정 문자로 시작하는 행을 찾을 때 자주 사용한다.

```python
cond = text_df['email'].str.contains('example1.co.kr', regex=False, na=False)
print(cond.sum())
print(text_df.loc[cond, ['reservation_id', 'email', 'status']].head().to_string(index=False))
```

### 6 문자열 길이와 도메인

#### 1. 전화번호 길이 확인

문자열 길이는 str.len( )으로 확인한다. 전화번호에서 하이픈을 제거한 뒤 길이를 확인하면 형식이 맞는 전화번호가 몇 개인지 빠르게 볼 수 있다. 국내 휴대전화 번호는 하이픈을 빼면 01012345678처럼 11자리이므로, 길이가 11이 아닌 값은 형식이 어긋난 후보로 볼 수 있다.

```python
print(text_df['phone_clean'].str.len().value_counts(dropna=False).sort_index().to_string())
print((text_df['phone_clean'].str.len() == 11).sum())
```

#### 2. 이메일 도메인 분리

이메일 주소는 항상 사용자명@도메인 구조이므로 @를 기준으로 나누면 원하는 부분만 뽑아낼 수 있다. str.split('@')은 각 값을 리스트로 쪼개고, 뒤에 붙는 .str[1]이 리스트의 두 번째 요소인 도메인만 가져온다.

```python
text_df['domain'] = text_df['email'].str.split('@').str[1]
print(text_df[['email', 'domain']].head().to_string(index=False))
print(text_df['domain'].value_counts().sort_index().to_string())
```

컬럼의 자료형은 dtype으로 확인합니다. object가 나오면 아직 문자라는 뜻이므로, pd.to_datetime( )으로 날짜형으로 바꿔 줍니다.

```python
import pandas as pd
df = pd.read_csv('data/part01_pandas/pandas_dates_strings.csv')
print(df['reserved_at'].dtype)
df['reserved_at'] = pd.to_datetime(df['reserved_at'])
print(df['reserved_at'].dtype)
print(df[['reservation_id', 'reserved_at']].head(3))
```

변환 전에는 object였던 자료형이 datetime64[ns]로 바뀐 것을 볼 수 있다. 이제 날짜형 데이터가 되었으므로 dt 접근자를 사용해 월, 일, 시간을 각각 추출하여 새 컬럼에 담으면 다음과 같다.

```python
df['reserve_month'] = df['reserved_at'].dt.month
df['reserve_day'] = df['reserved_at'].dt.day
df['reserve_hour'] = df['reserved_at'].dt.hour
print(df[['reservation_id', 'reserve_month', 'reserve_day', 'reserve_hour']].head())
```

월, 일, 시간이 숫자 컬럼으로 분리되었습니다. 이렇게 나눠 두면 '몇 월 예약이 많은지', '어느 시간대에 몰리는지' 같은 집계를 바로 할 수 있습니다. 날짜형 컬럼은 비교 연산도 가능하므로, 특정 날짜 이후의 예약만 골라낼 수도 있습니다.

```python
cond = df['reserved_at'] >= '2026-05-01'
print(cond.sum())
print(df[cond][['reservation_id', 'reserved_at', 'status']].head())
```
