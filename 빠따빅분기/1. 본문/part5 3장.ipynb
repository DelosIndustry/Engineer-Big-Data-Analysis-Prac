#### 2. 양측검정과 단측검정

### 2 검정 준비

#### 1. 공통 데이터와 검정 준비

주간 학습 시간의 평균이 기준값 40시간과 어떻게 다른지 확인한다. 이 데이터와 검정 결과는 다음 세 항목에서도 이어서 사용한다.

```python
study_path = (
    'data/part04_type3/one_sample_ttest_study_time/'
    'study_time.csv'
)
study = pd.read_csv(study_path)
base_value = 40
result_two = stats.ttest_1samp(
    study['study_hours'], base_value,
    alternative='two-sided'
)
result_greater = stats.ttest_1samp(
    study['study_hours'], base_value,
    alternative='greater'
)
result_less = stats.ttest_1samp(
    study['study_hours'], base_value,
    alternative='less'
)
print('shape: ', study.shape)
print('mean: ', format(study['study_hours'].mean(), '.4f'))
print('std: ', format(study['study_hours'].std(), '.4f'))
print('n: ', len(study))
print('two-sided statistic: ', format(result_two.statistic, '.6f'))
print('two-sided p-value: ', format(result_two.pvalue, '.6f'))
print('greater p-value: ', format(result_greater.pvalue, '.6f'))
print('less p-value: ', format(result_less.pvalue, '.6f'))
```

```python
alpha = 0.05
print('two-sided: ', result_two.pvalue < alpha)
print('greater: ', result_greater.pvalue < alpha)
print('less: ', result_less.pvalue < alpha)
```

## 05 t-검정 패턴

#### 1. 독립표본과 대응표본 데이터 불러오기

세 가지 t-검정을 각 데이터로 한 번에 실행해 결과를 비교해본다.

```python
app_path = (
    'data/part04_type3/independent_ttest_app_time/'
    'app_task_time.csv'
)
training_path = (
    'data/part04_type3/paired_ttest_training_score/'
    'training_score.csv'
)
app = pd.read_csv(app_path)
training = pd.read_csv(training_path)
group_a = app.loc[
    app['app_version'] == 'A', 'completion_minutes'
]
group_b = app.loc[
    app['app_version'] == 'B', 'completion_minutes'
]
levene_result = stats.levene(group_a, group_b)
ind_result = stats.ttest_ind(
    group_a, group_b,
    equal_var=False,
    alternative='two-sided'
)
paired_result = stats.ttest_rel(
    training['after_score'],
    training['before_score'],
    alternative='greater'
)
print('one_mean: ', format(study['study_hours'].mean(), '.4f'))
print('one_t: ', format(result_two.statistic, '.4f'))
print('one_p: ', format(result_two.pvalue, '.6f'))
print('A_mean: ', format(group_a.mean(), '.4f'))
print('B_mean: ', format(group_b.mean(), '.4f'))
print('levene_p: ', format(levene_result.pvalue, '.6f'))
print('ind_t: ', format(ind_result.statistic, '.4f'))
print('ind_p: ', format(ind_result.pvalue, '.6f'))
print('diff_mean: ', format(
    (training['after_score'] - training['before_score']).mean(), '.4f'
))
print('paired_t: ', format(paired_result.statistic, '.4f'))
print('paired_p: ', paired_result.pvalue)
```

## 06 비율 검정 패턴

### 1 데이터 준비

#### 1. 데이터 불러오기와 성공 수 확인

데이터를 불러와 성공 수와 가입 비율을 확인하고 단일 비율 검정까지 실행해본다.

```python
trial_path = (
    'data/part04_type3/logistic_regression_trial_signup/'
    'trial_signup.csv'
)
trial = pd.read_csv(trial_path)
success_count = trial['signed_up'].sum()
total_count = len(trial)
signup_rate = success_count/total_count
print('shape: ', trial.shape)
print('success_count: ', success_count)
print('total_count: ', total_count)
print('signup_rate: ', round(signup_rate, 4))
```

### 2 비율 검정 실행

#### 1. 단일 비율 검정

전체 가입 비율이 기준 비율 0.12와 다른지 statsmodels의 proportions_ztest( )로 양측검정한다. count에는 성공 수, nobs에는 전체 수, value에는 기준 비율을 넣는다.

```python
z_stat, p_value = proportions_ztest(
    count=success_count,
    nobs=total_count,
    value=0.12,
    alternative='two-sided'
)
print('z_stat: ', format(z_stat, '.4f'))
print('p_value: ', format(p_value, '.4f'))
```

#### 2. 두 집단 비율 차이 검정

집단별 요약을 만든 뒤 두 집단 비율 차이 검정을 실행해본다.

```python
summary = trial.groupby('email_opened')['signed_up'].agg(
    total='count',
    success='sum'
)
summary['rate'] = summary['success']/summary['total']
print(summary.round(4))
count = np.array([
    summary.loc[1, 'success'],
    summary.loc[0, 'success']
])
nobs = np.array([
    summary.loc[1, 'total'],
    summary.loc[0, 'total']
])
z_diff, p_diff = proportions_ztest(
    count=count,
    nobs=nobs,
    alternative='two-sided'
)
z_greater, p_greater = proportions_ztest(
    count=count,
    nobs=nobs,
    alternative='larger'
)
print('two_sample_z: ', round(z_diff, 4))
print('two_sample_pvalue: ', round(p_diff, 4))
print('one_sided_pvalue: ', round(p_greater, 4))
```

## 07 분산분석(ANOVA) 패턴

### 1 데이터와 가설

#### 1. 데이터와 가설 확인

데이터를 불러와 집단별 요약으로 가설을 확인해본다.

```python
support_path = (
    'data/part04_type3/anova_support_channel_rating/'
    'support_rating.csv'
)
support = pd.read_csv(support_path)
support_summary = support.groupby('support_channel')[
    'satisfaction'
].agg(['count', 'mean', 'std'])
print(support_summary.round(4))
```

### 2 분산분석 실행

#### 1. SciPy로 일원분산분석 수행하기

집단별 데이터를 리스트로 만들어 SciPy로 일원분산분석을 실행해본다.

```python
groups = [
    group['satisfaction'].values
    for _, group in support.groupby('support_channel')
]
anova_result = stats.f_oneway(*groups)
print('F: ', round(anova_result.statistic, 4))
print('p-value: ', anova_result.pvalue)
```

#### 2. statsmodels ANOVA 표 읽기

statsmodels로 같은 분석을 분산분석표 형태로 확인해본다.

```python
anova_model = ols(
    'satisfaction ~ C(support_channel)',
    data=support
).fit()
anova_table = sm.stats.anova_lm(anova_model, typ=2)
print(anova_table)
```

## 08 카이제곱 검정 패턴

### 1 검정 구분

#### 1. 적합도 검정과 독립성 검정

### 2 적합도·독립성 검정

#### 1. 데이터와 적합도 검정

데이터를 불러와 관측도수를 확인하고 적합도 검정을 실행해본다.

```python
device_path = (
    'data/part04_type3/chisquare_device_plan/'
    'device_plan.csv'
)
device = pd.read_csv(device_path)
observed = device['plan_grade'].value_counts().sort_index()
expected = np.repeat(len(device)/len(observed), len(observed))
goodness = stats.chisquare(observed, expected)
print(observed)
print('chi2: ', round(goodness.statistic, 4))
print('p-value: ', goodness.pvalue)
```

#### 2. 독립성 검정

교차표를 만들어 독립성 검정을 실행해본다.

```python
table = pd.crosstab(
    device['device_type'], device['plan_grade'])
chi2, p_value, dof, expected = stats.chi2_contingency(table)
print(table)
print('chi2: ', format(chi2, '.4f'))
print('p-value: ', format(p_value, '.6f'))
print('dof: ', dof)
print('expected: ')
print(np.round(expected, 4))
```

## 09 회귀분석 해석 패턴

### 1 모형 만들기와 값 확인

#### 1. 단순선형 회귀와 다중선형 회귀

데이터를 불러와 단순회귀와 다중회귀 모형을 함께 만들어 본다.

### 2 계수·지표 해석

#### 1. 회귀계수와 p-value 해석

#### 2. 결정계수(R²)와 F 검정

### 3 모형 재적합과 예측

#### 1. 유의 변수만 사용해 모형 다시 적합하기

변수 선별은 문제에서 유의한 변수만 사용하도록 명시한 경우에 적용한다. 초기 모형의 p-value에서 상수항을 제외하고 기준보다 작은 변수만 선택한 뒤 모형을 다시 적합한다.

```python
orders_refit = orders.copy()
orders_refit['date'] = pd.to_datetime(
    orders_refit['date']
)
orders_refit['weekday_num'] = (
    orders_refit['date'].dt.weekday
)
initial_X = sm.add_constant(
    orders_refit[
        ['ad_budget', 'is_weekend', 'weekday_num']
    ],
    has_constant='add'
)
initial_model = sm.OLS(
    orders_refit['orders'], initial_X
).fit()
significant = initial_model.pvalues.drop('const')
selected_cols = significant[
    significant < 0.05
].index.tolist()
refit_X = sm.add_constant(
    orders_refit[selected_cols],
    has_constant='add'
)
refit_model = sm.OLS(
    orders_refit['orders'], refit_X
).fit()
print('selected_cols: ', selected_cols)
print('refit_r2: ', round(refit_model.rsquared, 4))
```

#### 2. 새 관측치의 값 확인 후 예측하기

새 관측치에는 모형이 사용하는 모든 수치형 변수의 실젯값을 먼저 제공한다. 필수 컬럼이 누락되면 바로 오류를 발생시켜 입력 실수를 확인한다. reindex(fill_value=0)는 더미변수의 누락 열을 맞출 때 사용하는 방법이며, 필수 수치형 변수를 빠뜨린 상태를 숨기는 용도로 사용하지 않는다.

```python
new_data = pd.DataFrame({
    'ad_budget': [120.0],
    'is_weekend': [1]
})
missing_required = (
    set(selected_cols) - set(new_data.columns)
)
if missing_required:
    raise ValueError(
        f'필수 컬럼이 없습니다: {sorted(missing_required)}'
    )
new_X = sm.add_constant(
    new_data, has_constant='add'
)
new_X = new_X[refit_model.params.index]
new_prediction = refit_model.predict(new_X).iloc[0]
print('missing_required: ', missing_required)
print('new_prediction: ', round(new_prediction, 4))
```

#### 3. 날짜 순서로 나누어 예측 성능 확인하기

시간 순서를 반영한 평가에서는 날짜형으로 변환한 뒤 날짜순으로 정렬하고 학습용과 평가용 데이터를 나눈다. 다음 예제는 앞의 60행으로 학습하고 뒤의 20행으로 평가한다.

```python
eval_orders = orders.copy()
eval_orders['date'] = pd.to_datetime(
    eval_orders['date']
)
eval_orders['weekday_num'] = (
    eval_orders['date'].dt.weekday
)
eval_orders = (
    eval_orders
    .sort_values('date')
    .reset_index(drop=True)
)
train_orders = eval_orders.iloc[:60].copy()
test_orders = eval_orders.iloc[60:].copy()
base_cols = ['ad_budget', 'is_weekend', 'weekday_num']
train_X0 = sm.add_constant(
    train_orders[base_cols], has_constant='add'
)
train_model0 = sm.OLS(
    train_orders['orders'], train_X0
).fit()
selected_eval = train_model0.pvalues.drop('const')
selected_eval = selected_eval[
    selected_eval < 0.05
].index.tolist()
train_X = sm.add_constant(
    train_orders[selected_eval], has_constant='add'
)
train_model = sm.OLS(
    train_orders['orders'], train_X
).fit()
train_pred = train_model.predict(train_X)
train_corr = stats.pearsonr(
    train_orders['orders'], train_pred
)[0]
test_X = sm.add_constant(
    test_orders[selected_eval], has_constant='add'
)
test_X = test_X[train_model.params.index]
test_pred = train_model.predict(test_X)
test_rmse = np.sqrt(
    ((test_orders['orders'] - test_pred) ** 2).mean()
)
print('selected_cols: ', selected_eval)
print('train_corr: ', round(train_corr, 4))
print('test_rmse: ', round(test_rmse, 4))
```

## 10 로지스틱 회귀 해석 패턴

### 1 모형과 계수 해석

#### 1. 종속변수와 모형 만들기

5절에서 불러온 trial 데이터를 그대로 사용한다. signed_up은 0과 1로 구성된 종속변수이며, age·visited_pages·email_opened·previous_purchase를 독립변수로 사용한다.

```python
logit_X = trial[[
    'age',
    'visited_pages',
    'email_opened',
    'previous_purchase'
]].copy()
logit_X = sm.add_constant(logit_X)
logit_y = trial['signed_up']
logit_model = sm.Logit(logit_y, logit_X).fit(disp=False)
coef = logit_model.params
pvalues = logit_model.pvalues
odds_ratio = np.exp(coef)
pred_prob = logit_model.predict(logit_X)
print('coef: ')
print(coef.round(4))
print('pvalues: ')
print(pvalues)
print('odds_ratio: ')
print(odds_ratio.round(4))
print('pseudo_r2: ', round(logit_model.prsquared, 4))
print('pred_prob_head: ')
print(pred_prob.head().round(4))
```

#### 2. 회귀계수와 p-value

#### 3. 오즈비 해석

### 2 예측과 재적합

#### 1. 예측 확률과 분류값

statsmodels의 Logit 모형에서 model.predict(X)는 각 행이 signed_up=1이 될 예측 확률을 반환한다. 필요하면 기준값 0.5를 적용해 0과 1의 분류값을 만들 수 있다.

```python
pred_class = (pred_prob >= 0.5).astype(int)
print(pred_prob.head().round(4))
print(pred_class.head())
```

#### 2. 유의 변수만 사용해 다시 적합하기

범주형 변수를 포함한 로지스틱 회귀에서는 먼저 더미변수로 바꾼다. 유의 변수만 사용하도록 요구된 경우 초기 모형의 p-value를 기준으로 변수명을 선택해 다시 적합한다.

```python
logit_data = pd.get_dummies(
    trial[[
        'age',
        'visited_pages',
        'email_opened',
        'previous_purchase',
        'region'
    ]],
    columns=['region'],
    drop_first=True,
    dtype=int
)
initial_X = sm.add_constant(
    logit_data, has_constant='add'
)
initial_logit = sm.Logit(
    trial['signed_up'], initial_X
).fit(disp=False)
significant = initial_logit.pvalues.drop('const')
selected_cols = significant[
    significant < 0.05
].index.tolist()
refit_X = sm.add_constant(
    logit_data[selected_cols],
    has_constant='add'
)
refit_logit = sm.Logit(
    trial['signed_up'], refit_X
).fit(disp=False)
print('selected_cols: ', selected_cols)
print(np.exp(refit_logit.params).round(4))
```

### 3 오즈비와 예측 응용

#### 1. 여러 단위 증가의 오즈비 계산하기

회귀계수가 1단위 증가의 로그 오즈 변화량이라면 n단위 증가의 오즈비는 exp(계수 × n)으로 계산한다.

```python
five_page_odds = np.exp(
    refit_logit.params['visited_pages'] * 5
)
print(round(five_page_odds, 4))
```

#### 2. 범주값만 바꾸어 예측 확률 비교하기

다른 조건이 같은 두 프로필에서 email_opened만 0과 1로 바꾸면 이메일 열람 여부가 예측 확률에 반영되는 모습을 비교할 수 있다. 선택된 모든 수치형·더미변수 값을 명시한다.

```python
profiles = pd.DataFrame({
    'age': [35, 35],
    'visited_pages': [4, 4],
    'email_opened': [0, 1],
    'previous_purchase': [0, 0],
    'region_suburban': [0, 0]
})
profile_X = sm.add_constant(
    profiles, has_constant='add'
)
profile_X = profile_X[refit_logit.params.index]
profile_prob = refit_logit.predict(profile_X)
print(profile_prob.round(4))
print('difference: ', round(
    profile_prob.iloc[1] - profile_prob.iloc[0], 4
))
```
