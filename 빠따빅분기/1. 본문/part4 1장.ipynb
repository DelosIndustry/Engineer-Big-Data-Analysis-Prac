## 06 p-value

### 1 기준값에 따라 달라지는 p-value

기준값을 40과 42로 바꿔 가며 p-value가 어떻게 달라지는지 확인해본다.

```python
for standard in [40, 42]:
    result = stats.ttest_1samp(
        x,
        popmean=standard,
        alternative='two-sided'
    )
    print('standard: ', standard)
    print('t_statistic: ', round(result.statistic, 4))
    print('p_value: ', format(result.pvalue, '.6f'))
```

## 07 검정통계량

### 1 검정에 따라 달라지는 통계량 이름

### 2 t 검정통계량 직접 계산

단일표본 t 검정통계량은 다음 흐름으로 계산한다.t = (표본평균 - 기준값) / 표준오차

```python
standard = 40
sample_size = len(x)
sample_mean = x.mean()
sample_std = x.std()
standard_error = sample_std/math.sqrt(sample_size)
manual_t = (sample_mean-standard)/standard_error
result = stats.ttest_1samp(x, popmean=standard)
print('standard_error: ', round(standard_error, 4))
print('manual_t: ', round(manual_t, 4))
print('scipy_t_statistic: ', round(result.statistic, 4))
print('p_value: ', format(result.pvalue, '.6f'))
print('df: ', result.df)
```

## 08 신뢰구간

### 1 신뢰구간의 구성

### 2 95% 신뢰구간 계산

95% 신뢰구간을 공식대로 직접 계산하고 SciPy 결과와 비교해본다.

```python
confidence = 0.95
sample_size = len(x)
sample_mean = x.mean()
sample_std = x.std()
standard_error = sample_std/math.sqrt(sample_size)
df_value = sample_size-1
alpha = 1-confidence
t_critical = stats.t.ppf(1-alpha/2, df_value)
margin_error = t_critical*standard_error
ci_lower = sample_mean-margin_error
ci_upper = sample_mean+margin_error
ci = stats.t.interval(
    confidence,
    df_value,
    loc=sample_mean,
    scale=standard_error
)
print('t_critical: ', round(t_critical, 4))
print('margin_error: ', round(margin_error, 4))
print('ci_lower: ', round(ci_lower, 4))
print('ci_upper: ', round(ci_upper, 4))
print('scipy_ci: ', [round(float(v), 4) for v in ci])
print('contains_40: ', ci_lower <= 40 <= ci_upper)
```

## 09 단측검정과 양측검정

### 1 대립가설 방향과 alternative

### 2 세 방향의 p-value 비교

같은 데이터에서 세 방향의 p-value를 비교해본다.

```python
popmean = 40
alpha = 0.05
result_two = stats.ttest_1samp(x, popmean=popmean, alternative='two-sided')
result_greater = stats.ttest_1samp(x, popmean=popmean, alternative='greater')
result_less = stats.ttest_1samp(x, popmean=popmean, alternative='less')
print('t_statistic: ', round(result_two.statistic, 4))
print('two_sided_p: ', format(result_two.pvalue, '.6f'))
print('greater_p: ', format(result_greater.pvalue, '.6f'))
print('less_p: ', format(result_less.pvalue, '.6f'))
```

표본평균이 기준값 40보다 크므로 양측검정과 오른쪽 단측검정의 p-value는 작다. 반대 방향인 왼쪽 단측검정의 p-value는 0.999963으로 매우 크다.

```python
popmean = 42
for alt in ['two-sided', 'greater', 'less']:
    result = stats.ttest_1samp(x, popmean=popmean, alternative=alt)
    print(alt, format(result.pvalue, '.6f'))
```

### 2 답안 문장 만들기
