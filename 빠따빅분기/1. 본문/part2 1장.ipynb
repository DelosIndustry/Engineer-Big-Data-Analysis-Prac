다음 코드는 앞의 문제 문장을 그대로 풀이한 것이다.

```python
filtered = df[(df['flag'] == 1) & (df['group'] == 'G2')]
mean_amount = filtered['amount'].mean()
answer = int(mean_amount)
print(answer)
```

value_counts( )나 groupby( )를 사용하면 여러 값이 한 번에 출력된다. 이때 문제에서 그룹명을 묻는지, 해당 그룹의 빈도수를 묻는지 확인해야 한다.

```python
group_counts = df['group'].value_counts()
print(group_counts)
```

```python
print(group_counts.index[0])
print(group_counts.iloc[0])
```

## 02 정수 답안 처리와 출력값 확인

### 1 정수 답안 처리

#### 1. 문제에서 정수 처리 지시 확인하기

#### 2. int( )로 소수점 이하 제거하기

앞에서 사용한 조건에 맞는 amount 평균을 원래 값과 버림 처리한 값으로 비교한다. 현재 값은 양수이므로 int( )를 적용하면 소수점 이하가 제거되어 115865.75342465754가 115865가 된다. int( )는 일반적으로 값을 0 방향으로 절사하므로 음수에서는 수학적 내림과 결과가 다를 수 있다.

```python
cond = (df['flag'] == 1) & (df['group'] == 'G2')
mean_amount = df.loc[cond, 'amount'].mean()
print(mean_amount)
print(int(mean_amount))
```

#### 3. round( )로 반올림하기

같은 평균값을 이번에는 round( )로 반올림해본다.

```python
answer = round(mean_amount)
print(answer)
```

#### 4. 처음부터 정수로 계산되는 값

조건을 만족하는 행 수, 결측치 개수, 0과 1로 구성된 컬럼의 합계는 보통 처음부터 정수로 계산된다. 이 경우에는 별도로 int( )를 적용하지 않아도 된다.

```python
print((df['amount'] >= 150000).sum())
print(df['missing_value'].isna().sum())
print(df['flag'].sum())
```

### 2 출력값 확인

#### 1. 점검용 출력과 답안용 출력

#### 2. 중간 계산값 확인하기

조건이 올바르게 만들어졌는지 확인하려면 조건을 만족하는 행 수와 계산 결과를 함께 출력할 수 있다.

```python
cond = df['group'] == 'G1'
print(cond.sum())
print(df.loc[cond, 'score'].mean())
```

#### 3. answer 변수로 최종값 정리하기

문제 하나를 처음부터 끝까지 answer 변수로 정리해본다. G1 그룹의 score 평균을 소수 둘째 자리까지 반올림해 구하는 문제이다.

```python
cond = df['group'] == 'G1'
answer = round(df.loc[cond, 'score'].mean(), 2)
print(answer)
```

#### 4. 여러 줄 출력에서 답안 값 찾기

```python
print('flag가 1인 행 수: ', (df['flag'] == 1).sum())
print('missing_value 결측치 수: ', df['missing_value'].isna().sum())
```
