lecture_platform.csv 데이터는 course_title(강의명), instructor(강사명), open_date(개설일), status(모집상태), rating(강의평점)로 이루어져 있다. 다음 결과를 출력하시오.

**문제 1-1** course_title(강의명) 컬럼에 'Python' 또는 '파이썬'이 포함된 행의 개수를 출력하시오. 단, 영어의 경우 대소문자를 구분하지 않는다.

**문제 1-2** open_date(개설일)가 2020년 이후이면서, status(모집상태)가 '모집중'이고, rating(강의평점)이 4.0 이상인 강의의 개수를 출력하시오.

**제공 데이터 (lecture_platform.csv)**
| 컬럼명 | 설명 |
|---|---|
| course_title | 강의명 |
| instructor | 강사명 |

### 문제 풀이 방향

문제 1-1은 문자열 안에 특정 단어가 "포함"되어 있는지를 찾는 문제입니다. 지금까지 다뤘던 `str.startswith()`(첫 글자로 시작하는지)와 달리, 이번에는 문자열의 어느 위치에 있든 그 단어가 들어있기만 하면 되고, 두 개의 키워드('Python', '파이썬') 중 하나라도 있으면 되며, 영어는 대소문자까지 구분하지 않아야 합니다. 문제 1-2는 날짜에서 연도를 뽑아낸 뒤, 세 가지 조건을 동시에 만족하는 행을 찾는 문제입니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

````python
import pandas as pd

df = pd.read_csv('lecture_platform.csv')
print(df.shape)
df.head()
````

200개의 강의 데이터입니다. 3번 행의 강의명이 소문자 'python'으로 되어 있는 것처럼, 실제 데이터에는 대문자/소문자가 섞여 있을 수 있습니다.

**셀 2) 강의명에 'Python' 또는 '파이썬' 포함 여부 확인 (문제 1-1 정답)**

````python
mask = df['course_title'].str.contains('python|파이썬', case=False)
answer_1_1 = mask.sum()

print(mask.head(10))
print("1-1 정답:", answer_1_1)
````

`str.contains(패턴)`은 문자열 안에 그 패턴이 하나라도 포함되어 있는지를 True/False로 판별하는 함수입니다. `'python|파이썬'`처럼 파이프 기호(`|`)로 두 단어를 연결하면 "python 또는 파이썬"이라는 OR 조건을 표현할 수 있습니다. 이는 정규표현식(regular expression) 문법인데, `|`가 "또는"을 의미한다는 것만 알아두면 이런 형태의 조건은 대부분 해결할 수 있습니다.

`case=False`는 대소문자를 구분하지 않겠다는 옵션입니다. 이 옵션이 없으면 `'Python'`, `'PYTHoN'`, `'pythoN'`처럼 대소문자가 섞인 표기를 각각 다른 문자열로 취급해서 일부를 놓치게 됩니다. 문제에서 "영어의 경우 대소문자를 구분하지 않는다"고 명시했으므로 이 옵션을 반드시 챙겨야 합니다. `mask.sum()`으로 True(포함됨)인 행의 개수를 셌습니다.

> 💡 `str.contains()`는 특정 단어가 포함되어 있는지 찾을 때, `str.startswith()`는 특정 글자로 시작하는지 찾을 때 사용합니다. "어디에 있든 포함되면 된다"와 "반드시 맨 앞에 있어야 한다"는 서로 다른 조건이므로, 문제 지문에서 "포함"이라는 단어와 "~로 시작"이라는 단어를 구분해서 읽어야 합니다.

**셀 3) 개설일에서 연도 추출하기**

````python
df['year'] = pd.to_datetime(df['open_date']).dt.year
print(df[['open_date', 'year']].head())
````

`pd.to_datetime()`으로 문자열을 날짜 타입으로 바꾼 뒤, `.dt.year`로 연도만 뽑아내 새로운 컬럼으로 만들었습니다. 문자열 슬라이싱(`.str[:4]`)으로도 연도를 뽑아낼 수 있지만, 날짜 타입으로 변환해두면 연도뿐 아니라 월, 일, 요일 등 다양한 정보를 이후에도 쉽게 뽑아낼 수 있다는 장점이 있습니다.

**셀 4) 세 가지 조건을 동시에 만족하는 강의 개수 (문제 1-2 정답)**

````python
cond = (df['year'] >= 2020) & (df['status'] == '모집중') & (df['rating'] >= 4.0)
answer_1_2 = cond.sum()

print("1-2 정답:", answer_1_2)
````

세 개의 조건을 각각 괄호로 감싼 뒤 `&`(and)로 연결했습니다. 세 조건을 모두 만족하는 행만 True가 되고, `.sum()`으로 True인 행의 개수를 셌습니다.

### 전체 코드

````python
import pandas as pd

df = pd.read_csv('lecture_platform.csv')

# 1-1
mask = df['course_title'].str.contains('python|파이썬', case=False)
print("1-1 정답:", mask.sum())

# 1-2
df['year'] = pd.to_datetime(df['open_date']).dt.year
cond = (df['year'] >= 2020) & (df['status'] == '모집중') & (df['rating'] >= 4.0)
print("1-2 정답:", cond.sum())
````

### 정리

이 문제는 문자열 안에 특정 단어가 포함되어 있는지(`str.contains()`), 대소문자를 무시하는 방법(`case=False`), 그리고 여러 조건을 `&`로 동시에 만족시키는 방법을 함께 다뤘습니다. `str.contains()`에 파이프 기호로 여러 단어를 나열하면 OR 조건까지 표현할 수 있다는 것도 확인했습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

"포함(contains)"과 "정확히 일치(exact match)"는 비슷해 보이지만 결과가 크게 다를 수 있는 조건입니다.

**변형 예시: "포함"이 아니라 강의명이 정확히 "Python"과 일치(대소문자 무시)하는 행을 구하라면**

````python
exact_match = df['course_title'].str.lower() == 'python'
print("정확히 Python과 일치(대소문자 무시):", exact_match.sum())
````

`str.contains()`(포함) 결과는 42였지만, `str.lower() == 'python'`(정확히 일치)로 바꾸니 29로 줄어들었습니다. 그 차이(42-29=13)는 "Python 심화"처럼 'Python'이라는 단어를 포함하지만 강의명 전체가 정확히 'Python'과 같지는 않은 강의들입니다. `str.contains()`는 문자열 일부만 일치해도 True가 되지만, `==`(또는 `.str.lower()`와 결합한 비교)는 문자열 전체가 완전히 같아야 True가 된다는 차이를 꼭 기억해두어야 합니다.

## 문제 2. (10점)

store_sales.csv 데이터는 district(지역명), store(지점명), weekday(평일매출), weekend(주말매출)로 이루어져 있다. 다음 결과를 출력하시오.

**문제 2-1** 지역별로 전체매출 평균을 구한 뒤, 평균이 가장 높은 지역의 지점 개수와 평균이 두 번째로 높은 지역의 지점 개수를 더한 값을 출력하시오. 단, 전체매출은 평일매출과 주말매출의 합이며, 평일매출 컬럼의 결측치와 주말매출 컬럼의 결측치는 각각의 매출 컬럼 평균값으로 대체한다.

**문제 2-2** 전체매출의 평균이 가장 높은 지역과 평균이 두 번째로 높은 지역에 대한 두 그룹의 합동분산(Pooled Variance)을 출력하시오. 단, 결과는 반올림하여 소수점 셋째 자리까지 출력하고, 합동분산(s²ₚ)은 다음과 같이 구한다.

```
s²ₚ = ((n_A - 1)×s²_A + (n_B - 1)×s²_B) / (n_A + n_B - 2)
```
- s²ₚ: 합동분산(Pooled Variance)
- n_A, n_B: 각 그룹의 샘플 크기(데이터 개수)
- s²_A, s²_B: 각 그룹의 표본분산

**제공 데이터 (store_sales.csv)**
| 컬럼명 | 설명 |
|---|---|
| district | 지역명 |
| store | 지점명 |
| weekday | 평일매출 |
| weekend | 주말매출 |

### 문제 풀이 방향

이 문제는 두 컬럼(weekday, weekend)의 결측치를 각각 다른 값(그 컬럼 자신의 평균)으로 따로 채운 뒤, 두 컬럼을 더해 전체매출을 만드는 것부터 시작합니다. 그 다음 지역별 평균으로 순위를 매겨 1등, 2등 지역을 찾고, 문제 2-1은 그 두 지역에 속한 지점의 "개수"를, 문제 2-2는 그 두 지역 매출의 "합동분산"이라는 새로운 통계량을 계산합니다. 합동분산은 두 그룹의 분산을 각 그룹의 크기(자유도)로 가중 평균하여 하나로 합친 값으로, 공식이 주어져 있으므로 공식을 그대로 코드로 옮기면 됩니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

````python
import pandas as pd

df = pd.read_csv('store_sales.csv')
print(df.shape)
print(df.head())
print(df.isna().sum())
````

98개 지점의 매출 데이터이며, weekday에 12개, weekend에 9개의 결측치가 있습니다.

**셀 2) 결측치를 각 컬럼의 평균으로 대체하고 전체매출 계산하기**

````python
df['weekday'] = df['weekday'].fillna(df['weekday'].mean())
df['weekend'] = df['weekend'].fillna(df['weekend'].mean())
df['total'] = df['weekday'] + df['weekend']

print(df.isna().sum().sum())
df.head()
````

weekday의 결측치는 weekday 컬럼 자신의 평균으로, weekend의 결측치는 weekend 컬럼 자신의 평균으로 각각 따로 채웠습니다. 두 컬럼의 평균값 규모가 서로 다를 수 있으므로(주말 매출이 평일보다 대체로 높은 경향이 있다면), 서로 다른 컬럼의 평균을 섞어서 채우지 않도록 주의해야 합니다. 결측치를 다 채운 뒤에 두 컬럼을 더해 전체매출(total)을 만들었습니다.

**셀 3) 지역별 전체매출 평균 구하기**

````python
region_avg = df.groupby('district')['total'].mean().sort_values(ascending=False)
print(region_avg)
````

`groupby('district')['total'].mean()`으로 지역별 평균 매출을 구하고, 큰 순서대로 정렬했습니다. 강남구가 1등, 서초구가 2등인 것을 확인할 수 있습니다.

**셀 4) 1등, 2등 지역의 지점 개수 합 (문제 2-1 정답)**

````python
top1, top2 = region_avg.index[0], region_avg.index[1]
count1 = (df['district'] == top1).sum()
count2 = (df['district'] == top2).sum()
answer_2_1 = count1 + count2

print("1등 지역:", top1, "/ 지점수:", count1)
print("2등 지역:", top2, "/ 지점수:", count2)
print("2-1 정답:", answer_2_1)
````

`region_avg.index[0]`과 `[1]`로 1등, 2등 지역의 이름을 각각 가져온 뒤, `(df['district'] == 지역명).sum()`으로 그 지역에 속한 지점(행)이 몇 개인지 세었습니다. 여기서 주의할 점은, 매출 "평균"이 가장 높다고 해서 그 지역의 지점 "개수"도 가장 많은 것은 아니라는 것입니다. 실제로 2등 지역(서초구, 19개)이 1등 지역(강남구, 18개)보다 지점 수가 오히려 더 많습니다. 즉 평균 매출 순위와 지점 개수 순위는 서로 다른 기준이므로 혼동하지 않아야 합니다.

**셀 5) 1등, 2등 지역의 표본분산 계산하기**

````python
g1 = df[df['district'] == top1]['total']
g2 = df[df['district'] == top2]['total']
n1, n2 = len(g1), len(g2)
s1sq, s2sq = g1.var(ddof=1), g2.var(ddof=1)

print("n1:", n1, "n2:", n2)
print("s1^2:", s1sq, "s2^2:", s2sq)
````

각 지역에 속한 매출 값들만 따로 뽑아 표본분산을 구했습니다. `.var(ddof=1)`은 표본분산을 계산하는 pandas 함수로, `ddof=1`은 (n-1)로 나누는 표본분산 공식을 명시적으로 지정하는 옵션입니다. pandas의 `.var()`는 기본값 자체가 `ddof=1`이라 생략해도 같은 결과가 나오지만, 합동분산 공식에서 (n-1)이라는 표현이 명시적으로 나오는 만큼 이 옵션을 직접 써서 공식과 코드를 눈으로 대응시키기 쉽게 만들었습니다.

**셀 6) 합동분산 계산하기 (문제 2-2 정답)**

````python
pooled_var = ((n1-1)*s1sq + (n2-1)*s2sq) / (n1+n2-2)
answer_2_2 = round(pooled_var, 3)

print("2-2 정답:", answer_2_2)
````

문제에 주어진 공식 `s²ₚ = ((n_A-1)s²_A + (n_B-1)s²_B) / (n_A+n_B-2)`를 그대로 코드로 옮겼습니다. 이 공식은 두 그룹의 분산을 각 그룹의 자유도(n-1)만큼 가중치를 줘서 하나로 합친 값입니다. 그룹의 크기가 클수록(자유도가 클수록) 그 그룹의 분산이 합동분산 계산에 더 큰 비중으로 반영됩니다.

> 💡 통계 공식이 문제에 그림이나 수식으로 주어지는 경우, 공식의 각 기호(n_A, s²_A 등)가 코드의 어떤 변수와 대응하는지 먼저 눈으로 매칭해보는 것이 실수를 줄이는 방법입니다. 이번처럼 n1, n2, s1sq, s2sq라는 이름으로 미리 변수를 만들어두면, 공식을 옮겨적을 때 헷갈릴 가능성이 줄어듭니다.

### 전체 코드

````python
import pandas as pd

df = pd.read_csv('store_sales.csv')

df['weekday'] = df['weekday'].fillna(df['weekday'].mean())
df['weekend'] = df['weekend'].fillna(df['weekend'].mean())
df['total'] = df['weekday'] + df['weekend']

region_avg = df.groupby('district')['total'].mean().sort_values(ascending=False)
top1, top2 = region_avg.index[0], region_avg.index[1]

# 2-1
count1 = (df['district'] == top1).sum()
count2 = (df['district'] == top2).sum()
print("2-1 정답:", count1 + count2)

# 2-2
g1 = df[df['district'] == top1]['total']
g2 = df[df['district'] == top2]['total']
n1, n2 = len(g1), len(g2)
s1sq, s2sq = g1.var(ddof=1), g2.var(ddof=1)
pooled_var = ((n1-1)*s1sq + (n2-1)*s2sq) / (n1+n2-2)
print("2-2 정답:", round(pooled_var, 3))
````

### 정리

이 문제는 서로 다른 두 컬럼의 결측치를 각각 자신의 평균으로 채우는 것부터 시작해서, "평균이 높은 지역"과 "그 지역의 지점 개수"가 서로 다른 개념이라는 점, 그리고 통계 공식(합동분산)을 코드로 옮길 때는 공식의 기호와 코드 변수를 하나씩 대응시키는 것이 중요하다는 것을 확인했습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

"평균이 가장 높은 두 지역"이 아니라 "가장 낮은 두 지역"의 합동분산을 구하라는 변형이 나올 수 있습니다.

**변형 예시: 평균이 가장 낮은 두 지역의 합동분산을 구하라면**


````python
bottom1, bottom2 = region_avg.index[-1], region_avg.index[-2]
print("최하위 지역:", bottom1, "/ 차하위 지역:", bottom2)

gb1 = df[df['district'] == bottom1]['total']
gb2 = df[df['district'] == bottom2]['total']
nb1, nb2 = len(gb1), len(gb2)
sb1sq, sb2sq = gb1.var(ddof=1), gb2.var(ddof=1)

pooled_var_bottom = ((nb1-1)*sb1sq + (nb2-1)*sb2sq) / (nb1+nb2-2)
print("하위 2개 지역 합동분산:", round(pooled_var_bottom, 3))
````

`region_avg.index[0]`, `[1]`(정렬된 결과의 맨 앞 두 개, 즉 1등·2등)이 `region_avg.index[-1]`, `[-2]`(정렬된 결과의 맨 뒤 두 개, 즉 꼴등·차꼴등)로 바뀌었을 뿐, 나머지 계산 과정은 완전히 동일합니다. 파이썬에서 음수 인덱스(`-1`, `-2`)는 뒤에서부터 세는 위치를 가리킨다는 것을 활용하면, 정렬된 결과의 양쪽 끝을 모두 쉽게 다룰 수 있습니다.

## 문제 3. (10점)

**문제 3-1** 자격증을 취득하지 않은 직원들의 나이의 평균값을 출력하시오. 그리고 자격증을 취득한 직원들에 대해 평균 취득 개수에 대한 자격증을 취득한 직원들의 평균 나이(=(평균 나이)/(평균 취득 개수))를 출력하시오.

**문제 3-2** 부서별로 미취득 직원 수와 취득 직원 수의 차이(절댓값)를 구하고 이 값이 가장 큰 부서와 그 값을 출력하시오.

**제공 데이터 (cert_status.csv)**
| 컬럼명 | 설명 |
|---|---|
| id | 사번 |
| dept | 부서 |
| field | 직무 |
| phone | 전화번호 |
| age | 나이 |
| type | 자격증 취득 분류 (0: 미취득, 0보다 큰 수: 취득 개수) |

### 문제 풀이 방향

문제 3-1은 미취득자와 취득자를 조건으로 나눈 뒤, 취득자 그룹에 대해서는 "평균 나이를 평균 취득 개수로 나눈 값"이라는 파생 계산까지 요구합니다. 값 하나를 그대로 구하는 것이 아니라, 이미 구한 두 개의 평균값을 다시 조합해야 한다는 점에서 지금까지와는 다른 유형입니다. 문제 3-2는 부서별로 두 조건(미취득/취득)에 해당하는 인원수를 각각 센 뒤 그 차이를 비교하는 문제로, 이전에 다룬 "그룹 간 합계 차이" 패턴과 뼈대는 비슷하지만 합계가 아니라 인원수(개수)를 비교한다는 점이 다릅니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

````python
import pandas as pd

df = pd.read_csv('cert_status.csv')
print(df.shape)
df.head()
````

320명의 직원 데이터이며, type이 0이면 자격증 미취득, 0보다 크면 취득한 자격증 개수를 의미합니다.

**셀 2) 미취득자의 평균 나이 계산 (A값)**

````python
no_cert = df[df['type'] == 0]
answer_A = round(no_cert['age'].mean(), 3)

print("미취득자 수:", len(no_cert))
print("A(미취득자 평균나이):", answer_A)
````

`df[df['type'] == 0]`으로 미취득자(type이 정확히 0인 행)만 걸러낸 뒤, 그들의 나이 평균을 구했습니다.

**셀 3) 취득자의 평균 나이와 평균 취득 개수 계산하기**

````python
has_cert = df[df['type'] > 0]
avg_age_has = has_cert['age'].mean()
avg_type_has = has_cert['type'].mean()

print("취득자 평균나이:", avg_age_has)
print("취득자 평균 취득개수:", avg_type_has)
````

`df[df['type'] > 0]`으로 이번엔 자격증을 하나라도 취득한 직원들만 걸러냈습니다. 이 그룹에 대해 나이의 평균과, type(취득 개수) 자체의 평균을 각각 구했습니다. type이 1, 2, 3처럼 자격증 개수를 나타내는 숫자이므로, 이 값의 평균을 구하면 "취득자들이 평균적으로 몇 개의 자격증을 가지고 있는지"를 알 수 있습니다.

**셀 4) B값 계산하기 (문제 3-1 정답)**

````python
answer_B = round(avg_age_has / avg_type_has, 3)
print("B:", answer_B)
````

문제에서 요구한 대로 "평균 나이를 평균 취득 개수로 나눈 값"을 계산했습니다. 이 값 자체는 물리적으로 뚜렷한 의미를 갖는 지표는 아니고, 문제에서 정의한 대로 두 평균값을 조합해서 만든 파생 지표입니다. 이렇게 문제에서 "=(A)/(B)"처럼 계산식을 직접 정의해주는 경우에는, 그 정의를 있는 그대로 따르는 것이 중요하며 별도의 통계적 의미를 스스로 해석하려 하지 않아도 됩니다.

**셀 5) 부서별 미취득자/취득자 인원수 계산하기**

````python
dept_no = df[df['type'] == 0].groupby('dept').size()
dept_has = df[df['type'] > 0].groupby('dept').size()

print("부서별 미취득자 수:\n", dept_no)
print("부서별 취득자 수:\n", dept_has)
````

미취득자와 취득자를 각각 조건으로 걸러낸 뒤, `groupby('dept').size()`로 부서별 인원수를 세었습니다. `.size()`는 그룹별로 행이 몇 개 있는지 세어주는 함수로, 특정 컬럼의 값을 집계하는 `.sum()`이나 `.mean()`과 달리 컬럼 지정 없이 그룹 자체의 크기(행 개수)를 바로 알려줍니다.

**셀 6) 인원수 차이가 가장 큰 부서 찾기 (문제 3-2 정답)**

````python
diff = (dept_no - dept_has).abs().sort_values(ascending=False)
print(diff)

max_dept = diff.idxmax()
max_value = diff.max()
print("최대 부서:", max_dept, "/ 값:", max_value)
````

`dept_no - dept_has`는 두 Series를 부서(인덱스)가 같은 것끼리 자동으로 짝지어 뺄셈합니다. 두 Series 모두 인덱스가 부서명으로 되어 있기 때문에, 순서가 다르더라도 부서명이 일치하는 값끼리 정확히 계산됩니다. 여기에 `.abs()`로 절댓값을 취해 방향(어느 쪽이 더 많은지)과 무관하게 차이의 크기만 비교했습니다. 영업본부는 미취득자(73명)가 취득자(29명)보다 훨씬 많아 그 차이가 44명으로 가장 컸습니다.

### 전체 코드

````python
import pandas as pd

df = pd.read_csv('cert_status.csv')

# 3-1
no_cert = df[df['type'] == 0]
has_cert = df[df['type'] > 0]
answer_A = round(no_cert['age'].mean(), 3)
answer_B = round(has_cert['age'].mean() / has_cert['type'].mean(), 3)
print("A:", answer_A)
print("B:", answer_B)

# 3-2
dept_no = df[df['type'] == 0].groupby('dept').size()
dept_has = df[df['type'] > 0].groupby('dept').size()
diff = (dept_no - dept_has).abs()
print("최대 부서:", diff.idxmax(), "/ 값:", diff.max())
````

### 정리

이 문제는 조건으로 그룹을 나눈 뒤 평균을 구하는 것까지는 익숙한 흐름이었지만, 그렇게 구한 평균값들을 문제에서 정의한 계산식(평균나이/평균취득개수)에 따라 다시 조합해야 한다는 점이 새로웠습니다. 또한 `groupby().size()`로 그룹별 인원수(개수)를 세고, 두 그룹의 인원수 차이를 비교하는 것은 지금까지 다룬 "그룹별 합계(sum) 차이"와 형태는 비슷하지만 세는 대상이 값이 아니라 행의 개수라는 점에서 구분해야 합니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

"차이가 가장 큰" 부서가 아니라 "차이가 가장 작은" 부서를 물어보는 방식으로 변형될 수 있습니다.

**변형 예시: 인원수 차이가 가장 작은(가장 균형 잡힌) 부서를 구하라면**

````python
diff_sorted = diff.sort_values()
min_dept = diff_sorted.index[0]
min_value = diff_sorted.iloc[0]

print(diff_sorted)
print("차이가 가장 작은 부서:", min_dept, "/ 값:", min_value)
````

`sort_values()`를 오름차순(기본값) 그대로 사용하면 가장 작은 값이 맨 위로 오므로, `.index[0]`과 `.iloc[0]`으로 차이가 가장 작은 부서와 그 값을 바로 조회할 수 있습니다. 재무본부는 미취득자와 취득자의 수가 비교적 비슷해서(23명 vs 27명) 차이가 4명으로 가장 작았는데, 이는 영업본부처럼 한쪽으로 크게 치우치지 않고 두 그룹이 상대적으로 균형을 이루고 있다는 뜻으로 해석할 수 있습니다.

## 이번 회차에서 사용한 핵심 함수 정리
| 상황 | 함수 | 핵심 이유 |
|---|---|---|
| 문자열에 특정 단어가 포함되어 있는지 확인 | `str.contains('단어1\|단어2', case=False)` | 파이프(\|)로 OR 조건, case=False로 대소문자 무시 |
| 포함 vs 정확히 일치 구분 | `str.contains()` vs `== '값'` | 부분 일치와 완전 일치는 결과가 다를 수 있음 |
| 날짜에서 연도만 추출 | `pd.to_datetime(컬럼).dt.year` | 날짜 타입으로 변환하면 연/월/일 등을 자유롭게 추출 가능 |
| 여러 조건을 동시에 만족하는 행 찾기 | `(조건1) & (조건2) & (조건3)` | 각 조건을 괄호로 감싸고 &로 연결 |
| 서로 다른 컬럼의 결측치를 각각 대체 | `df[컬럼].fillna(df[컬럼].mean())`을 컬럼별로 각각 적용 | 컬럼마다 값의 규모가 다를 수 있어 다른 컬럼 평균을 섞으면 안 됨 |
| 표본분산 계산 | `.var(ddof=1)` | ddof=1은 (n-1)로 나누는 표본분산, pandas 기본값과 동일 |
| 통계 공식을 코드로 옮기기 | 공식의 기호와 변수명을 미리 대응시켜 정의 | 복잡한 수식일수록 기호-변수 매칭이 실수를 줄임 |
| 그룹별 행 개수(인원수) 세기 | `groupby(기준).size()` | 특정 컬럼의 합계가 아니라 그룹 자체의 크기(개수)가 필요할 때 |
| 인덱스가 같은 두 Series의 뺄셈 | `Series1 - Series2` | 인덱스(부서명 등) 기준으로 자동 정렬되어 계산됨 |
