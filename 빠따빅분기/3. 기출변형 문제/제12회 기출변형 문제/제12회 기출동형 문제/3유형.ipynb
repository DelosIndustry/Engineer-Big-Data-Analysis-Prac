## 문제 1. (10점) — 가설검정

coaching_score.csv 데이터는 id(번호), pretest(사전평가 점수), posttest(사후평가 점수), coaching(코칭 프로그램 참여여부)로 이루어져 있으며, coaching=1이면 프로그램 참여, 0이면 프로그램 미참여를 나타낸다. 다음 수행결과를 출력하시오. 단, 모든 결과는 반올림하여 소수점 셋째 자리까지 출력하시오.

**문제 1-1** 사전평가 점수(pretest)의 평균이 90점과 차이 있는지를 검정하고 검정통계량을 이용하여 유의확률(p-value)을 출력하시오.

**문제 1-2** 코칭 프로그램 참여로 사후평가 점수가 사전평가 점수에 비해 향상되었는지를 검정하고 검정통계량과 유의확률(p-value)을 출력하시오.

**문제 1-3** A그룹(코칭 프로그램 참여)과 B그룹(코칭 프로그램 미참여)에 대해 사후평가 평균 점수와 사전평가 평균 점수의 차이에 대한 95% 신뢰구간의 하한값을 출력하시오. 즉, 코칭 프로그램에 참여한 집단(A그룹)과 참여하지 않은 집단(B그룹) 사이의, 사후와 사전 점수 차이에 대한 신뢰구간의 하한값을 출력하시오.

**제공 데이터 (coaching_score.csv)**
| 컬럼명 | 설명 |
|---|---|
| id | 참가자 번호 |
| pretest | 사전평가 점수 |
| posttest | 사후평가 점수 |
| coaching | 코칭 프로그램 참여여부 (1: 참여, 0: 미참여) |

### 문제 풀이 방향

이 문제는 세 개의 소문항이 각각 다른 통계 개념을 요구합니다. 1-1은 데이터 전체를 대상으로 한 단일표본 t검정이고, 1-2는 데이터 전체가 아니라 코칭에 참여한 사람들(coaching=1)만 골라내서 그 사람들의 사전-사후 점수를 비교하는 대응표본 t검정입니다. 1-3은 지금까지 다룬 적 없는 개념인 신뢰구간(Confidence Interval)을 직접 계산해야 합니다. 지금까지는 p-value로 "유의한가 아닌가"만 판단했지만, 신뢰구간은 "차이가 어느 범위 안에 있을 것으로 추정되는가"를 구체적인 숫자 범위로 알려준다는 점이 다릅니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

```python
import pandas as pd

df = pd.read_csv('coaching_score.csv')
print(df.shape)
df.head()
```

120명의 참가자 데이터입니다.

**셀 2) 사전평가 점수에 대한 단일표본 t검정 수행**

```python
from scipy.stats import ttest_1samp

t1, p1 = ttest_1samp(df['pretest'], 90)
print("t통계량:", t1)
print("p-value:", p1)
```

`ttest_1samp(데이터, 비교값)`으로 사전평가 점수 전체의 평균이 90점과 같다고 볼 수 있는지 검정했습니다. 귀무가설은 "사전평가 평균이 90점이다"입니다.

**셀 3) 검정통계량과 p-value 출력 (문제 1-1 정답)**

```python
answer_1_1_t = round(t1, 3)
answer_1_1_p = round(p1, 3)

print("1-1 검정통계량:", answer_1_1_t, "/ p-value:", answer_1_1_p)
```

t통계량이 음수라는 것은 실제 평균(약 86.5점)이 비교 기준(90점)보다 낮다는 뜻입니다. p-value가 0.05보다 훨씬 작으므로, "평균이 90점이다"라는 귀무가설을 기각합니다.

**셀 4) 코칭 참여 그룹만 골라내기**

```python
grpA = df[df['coaching'] == 1]
print(grpA.shape)
print(grpA.head())
```

`df[df['coaching'] == 1]`로 전체 120명 중 코칭 프로그램에 참여한 67명만 걸러냈습니다. 문제 1-2는 전체 데이터가 아니라 이 부분집단만을 대상으로 하는 검정이라는 점이 중요합니다.

**셀 5) 참여 그룹의 사전-사후 점수 대응표본 t검정**

```python
from scipy.stats import ttest_rel

t2, p2 = ttest_rel(grpA['posttest'], grpA['pretest'])
print("t통계량:", t2)
print("p-value:", p2)
```

`ttest_rel(a, b)`는 같은 대상을 두 번 측정한 값(사후, 사전)의 차이를 검정하는 대응표본 t검정입니다. 여기서는 코칭에 참여한 67명만 대상으로, 사후평가가 사전평가보다 높아졌는지를 검정했습니다. 문제에서 "향상되었는지"를 물었으므로 `ttest_rel(posttest, pretest)`처럼 사후를 먼저 넣어, 양수 t통계량이 나오면 향상되었다는 방향과 일치하도록 순서를 맞췄습니다.

**셀 6) 검정통계량과 p-value 출력 (문제 1-2 정답)**

```python
answer_1_2_t = round(t2, 3)
answer_1_2_p = round(p2, 3)

print("1-2 검정통계량:", answer_1_2_t, "/ p-value:", answer_1_2_p)
```

t통계량이 양수(9.581)로 크게 나왔고 p-value가 매우 작으므로, 코칭 프로그램에 참여한 사람들의 사후평가 점수가 사전평가보다 통계적으로 유의하게 향상되었다고 결론 내릴 수 있습니다.

**셀 7) 그룹별 점수 차이(diff) 계산하기**

```python
df['diff'] = df['posttest'] - df['pretest']
diffA = df[df['coaching'] == 1]['diff']
diffB = df[df['coaching'] == 0]['diff']

print(diffA.describe())
print(diffB.describe())
```

전체 데이터에 대해 "사후-사전" 점수 차이(diff)를 먼저 계산한 뒤, 코칭 참여 그룹(diffA)과 미참여 그룹(diffB)으로 나눠서 각각의 분포를 확인했습니다. 참여 그룹(diffA)의 평균 향상 점수(4.91점)가 미참여 그룹(diffB)의 평균 향상 점수(1.25점)보다 뚜렷하게 높다는 것을 미리 확인할 수 있습니다.

**셀 8) 두 그룹 차이의 95% 신뢰구간 계산 (문제 1-3 정답)**

```python
from statsmodels.stats.weightstats import DescrStatsW, CompareMeans

cm = CompareMeans(DescrStatsW(diffA), DescrStatsW(diffB))
ci_low, ci_high = cm.tconfint_diff(alpha=0.05, usevar='pooled')

print("95% 신뢰구간:", ci_low, ci_high)
answer_1_3 = round(ci_low, 3)
print("1-3 정답(하한값):", answer_1_3)
```

지금까지는 p-value로 "유의한가 아닌가"만 확인했지만, 신뢰구간은 "차이가 실제로 어느 범위 안에 있을 것으로 추정되는가"를 구간(하한값~상한값)으로 보여줍니다. `DescrStatsW(데이터)`는 하나의 그룹에 대한 기술통계 정보를 담는 객체이고, `CompareMeans(그룹A정보, 그룹B정보)`는 두 그룹을 비교하기 위한 객체입니다. `.tconfint_diff(alpha=0.05, usevar='pooled')`는 t분포를 이용해 두 그룹 평균 차이의 신뢰구간을 계산해주는 함수로, `alpha=0.05`는 95% 신뢰수준(1-0.05)을, `usevar='pooled'`는 두 그룹의 분산이 같다고 가정하는(합동분산을 사용하는) 방식을 의미합니다.

계산된 신뢰구간이 (2.086, 5.245)로 나왔는데, 이 구간이 0을 포함하지 않고 완전히 양수 영역에 있다는 것은, "참여 그룹의 향상 점수가 미참여 그룹보다 통계적으로 유의하게 크다"는 것을 다시 한번 뒷받침합니다. 만약 신뢰구간에 0이 포함되어 있었다면, "두 그룹의 차이가 없을 가능성도 배제할 수 없다"는 뜻이 됩니다.

> 💡 신뢰구간이 0을 포함하는지 여부는 p-value가 유의수준보다 작은지를 판단하는 것과 사실상 같은 정보를 담고 있습니다. 신뢰구간이 0을 포함하지 않으면 그 차이는 유의수준에서 통계적으로 유의하다고 해석할 수 있고, 반대로 0을 포함하면 유의하지 않다고 볼 수 있습니다.

### 전체 코드

```python
import pandas as pd
from scipy.stats import ttest_1samp, ttest_rel
from statsmodels.stats.weightstats import DescrStatsW, CompareMeans

df = pd.read_csv('coaching_score.csv')

# 1-1: 단일표본 t검정
t1, p1 = ttest_1samp(df['pretest'], 90)
print("1-1 검정통계량:", round(t1, 3), "/ p-value:", round(p1, 3))

# 1-2: 참여 그룹만 대응표본 t검정
grpA = df[df['coaching'] == 1]
t2, p2 = ttest_rel(grpA['posttest'], grpA['pretest'])
print("1-2 검정통계량:", round(t2, 3), "/ p-value:", round(p2, 3))

# 1-3: 두 그룹 차이의 95% 신뢰구간
df['diff'] = df['posttest'] - df['pretest']
diffA = df[df['coaching'] == 1]['diff']
diffB = df[df['coaching'] == 0]['diff']
cm = CompareMeans(DescrStatsW(diffA), DescrStatsW(diffB))
ci_low, ci_high = cm.tconfint_diff(alpha=0.05, usevar='pooled')
print("1-3 정답(하한값):", round(ci_low, 3))
```

### 정리

이 문제는 세 가지 검정 개념(단일표본, 대응표본, 신뢰구간)을 하나의 데이터로 순서대로 다뤘습니다. 특히 문제 1-2처럼 "전체가 아니라 특정 조건을 만족하는 부분집단만" 검정 대상으로 삼아야 하는 경우가 있다는 것과, p-value 대신 신뢰구간을 구하는 방법(`CompareMeans().tconfint_diff()`)을 새로 익혔습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

신뢰구간은 하한값뿐 아니라 상한값을 물어볼 수도 있습니다.

**변형 예시: 하한값이 아니라 상한값을 구하라면**

```python
answer_upper = round(ci_high, 3)
print("95% 신뢰구간 상한값:", answer_upper)
```

`tconfint_diff()`가 돌려주는 두 값(`ci_low`, `ci_high`) 중 이미 계산해둔 `ci_high`를 그대로 반올림하기만 하면 됩니다. 신뢰구간은 항상 (하한값, 상한값) 한 쌍으로 계산되므로, 문제에서 어느 쪽을 요구하는지만 정확히 확인하면 이미 구한 결과에서 값만 바꿔 꺼내면 됩니다.

## 문제 2. (10점) — 로지스틱 회귀분석

lms_completion.csv 데이터는 id(번호), login_freq(주당 접속빈도), assignment_submit(과제제출 건수), forum_post(게시글 수), quiz_score(퀴즈점수), study_time(학습시간), prob(수료확률)로 이루어져 있다. (login_freq, assignment_submit, forum_post, quiz_score, study_time)을 이용하여 수료 여부를 예측하기 위해 로지스틱 회귀분석 모형을 이용하고자 한다. 다음 수행결과를 출력하시오. 단, 모든 결과는 반올림하여 소수점 셋째 자리까지 출력하시오.

**문제 2-1** study_time(학습시간)에 대한 수료확률의 오즈비(Odds Ratio)를 출력하시오.

**문제 2-2** 다음 데이터(newdata)에 대한 예측 확률값을 출력하시오. newdata: (login_freq, assignment_submit, forum_post, quiz_score, study_time) = (15, 13, 6, 8, 7).

**문제 2-3** prob(수료확률) 값이 0.5 이상이면 1(수료), 0.5 미만이면 0(미수료)으로 정의할 때 실제값과 예측값을 비교하여 정확도를 출력하시오.

**제공 데이터 (lms_completion.csv)**
| 컬럼명 | 설명 |
|---|---|
| id | 학습자 번호 |
| login_freq | 주당 접속빈도 |
| assignment_submit | 과제제출 건수 |
| forum_post | 게시글 수 |
| quiz_score | 퀴즈점수 |
| study_time | 학습시간 |
| prob | 수료확률 (연속값) |

### 문제 풀이 방향

이 문제에서 가장 먼저 주목해야 할 점은, prob 컬럼이 이미 0.076, 0.355처럼 연속적인 확률값으로 주어져 있다는 것입니다. 로지스틱 회귀분석은 원래 0 또는 1의 이진 종속변수를 예측하기 위한 모델인데, 이 데이터에는 그런 이진값이 없습니다. 그래서 본격적인 모델링에 앞서, prob 값을 0.5를 기준으로 0 또는 1로 변환해서 새로운 이진 종속변수를 먼저 만들어야 합니다. 이렇게 만든 이진변수를 종속변수로 삼아 로지스틱 회귀를 적합한 뒤, 오즈비 해석과 새 데이터 예측, 그리고 정확도 계산까지 진행하겠습니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

```python
import pandas as pd

df = pd.read_csv('lms_completion.csv')
print(df.shape)
df.head()
```

500명의 학습자 데이터이며, prob 컬럼이 0에서 1 사이의 연속된 확률값으로 주어져 있습니다.


**셀 2) prob를 기준으로 이진 종속변수(completed) 생성하기**

```python
df['completed'] = (df['prob'] >= 0.5).astype(int)

print(df[['prob', 'completed']].head(10))
print(df['completed'].value_counts())
```

`df['prob'] >= 0.5`는 각 행의 prob 값이 0.5 이상인지를 True/False로 판별하고, `.astype(int)`로 이를 1/0으로 바꿨습니다. 이렇게 만든 `completed`가 앞으로 로지스틱 회귀의 종속변수(그리고 나중에 정확도 계산 시 "실제값")로 사용됩니다. 500명 중 104명(약 20.8%)이 수료(1)로 분류되었습니다.

**셀 3) 로지스틱 회귀모형 적합**

```python
import statsmodels.api as sm

X = df.drop(columns=['id', 'prob', 'completed'])
y = df['completed']
X = sm.add_constant(X)

model = sm.Logit(y, X).fit()
print(model.summary())
```

`X`에서 id(식별자), prob(원본 연속값), completed(방금 만든 종속변수 자신)를 모두 제외했습니다. 특히 prob는 completed를 만드는 데 사용한 원천 데이터이므로, 이를 독립변수에 그대로 남겨두면 "답이 답을 예측하는" 상황이 되어버립니다. quiz_score는 p-value가 0.485로 유의하지 않지만, 이 문제에서는 유의한 변수만 재적합하라는 지시가 없으므로 모든 변수를 그대로 유지한 모형을 사용합니다.

**셀 4) study_time의 오즈비 계산 (문제 2-1 정답)**

```python
import numpy as np

odds_time = np.exp(model.params['study_time'])
print("2-1 정답:", round(odds_time, 3))
```

study_time의 계수(0.1558)에 지수를 취해 오즈비를 구했습니다. 학습시간이 1단위 늘어날 때마다 수료 오즈가 약 1.169배 높아진다는 뜻입니다.

**셀 5) 새로운 데이터의 예측 확률 (문제 2-2 정답)**

```python
new_data = pd.DataFrame({'const': [1], 'login_freq': [15], 'assignment_submit': [13],
                          'forum_post': [6], 'quiz_score': [8], 'study_time': [7]})
print(new_data)

pred_new = model.predict(new_data)
print("2-2 정답:", round(pred_new.values[0], 3))
```

새로운 데이터를 데이터프레임으로 만들 때, `const`를 포함해 모형에 사용된 컬럼과 정확히 같은 이름과 순서로 구성했습니다. 예측 확률이 약 0.334로, 0.5보다 낮아 이 조건의 학습자는 수료보다는 미수료로 예측된다는 것을 알 수 있습니다.

**셀 6) 전체 데이터에 대한 예측 클래스 계산**

```python
pred_all_prob = model.predict(X)
pred_class = (pred_all_prob >= 0.5).astype(int)

print(pred_class.head())
```

모형으로 500명 전체의 수료 확률을 예측한 뒤, 0.5를 기준으로 다시 0 또는 1로 분류했습니다. 이 `pred_class`가 모형의 최종 판단(예측값)이 됩니다.

**셀 7) 정확도 계산 (문제 2-3 정답)**

```python
accuracy = (pred_class == y).mean()
print("2-3 정답:", round(accuracy, 3))
```

`pred_class == y`는 예측값(pred_class)과 실제값(y, 즉 completed)이 같은지를 행마다 True/False로 판별합니다. True는 1, False는 0으로 취급되므로, `.mean()`을 취하면 "전체 중 예측이 실제와 일치한 비율", 즉 정확도(Accuracy)가 됩니다. 지금까지 다뤄온 민감도·특이도는 각각 실제 양성/음성 중에서의 적중률만 따로 보는 지표였다면, 정확도는 양성과 음성을 구분하지 않고 전체 예측 중 맞춘 비율을 하나의 숫자로 나타낸다는 점이 다릅니다.

> 💡 정확도는 계산이 직관적이지만, 클래스 비율이 한쪽으로 크게 치우친 데이터(이번처럼 미수료가 396명, 수료가 104명)에서는 주의가 필요합니다. 만약 모형이 무조건 "미수료(0)"라고만 예측해도 정확도는 396/500=79.2%가 나오는데, 이는 지금 구한 정확도(79.2%)와 거의 같은 수준입니다. 즉 정확도 하나만으로는 이 모형이 실제로 수료자를 잘 가려내는지 판단하기 어려울 수 있으므로, 클래스 비율이 불균형한 데이터에서는 민감도·특이도 같은 지표를 함께 살펴보는 것이 안전합니다.

### 전체 코드

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

df = pd.read_csv('lms_completion.csv')

# prob를 기준으로 이진 종속변수 생성
df['completed'] = (df['prob'] >= 0.5).astype(int)

X = df.drop(columns=['id', 'prob', 'completed'])
y = df['completed']
X = sm.add_constant(X)

model = sm.Logit(y, X).fit()

# 2-1: study_time 오즈비
odds_time = round(np.exp(model.params['study_time']), 3)
print("2-1 정답:", odds_time)

# 2-2: 새 데이터 예측 확률
new_data = pd.DataFrame({'const': [1], 'login_freq': [15], 'assignment_submit': [13],
                          'forum_post': [6], 'quiz_score': [8], 'study_time': [7]})
pred_new = round(model.predict(new_data).values[0], 3)
print("2-2 정답:", pred_new)

# 2-3: 정확도
pred_class = (model.predict(X) >= 0.5).astype(int)
accuracy = round((pred_class == y).mean(), 3)
print("2-3 정답:", accuracy)
```

### 정리

이 문제의 핵심은 연속형 확률 데이터를 이진 종속변수로 먼저 변환하는 전처리 단계였습니다. 이 변환을 빠뜨리면 로지스틱 회귀 자체를 적용할 수 없습니다. 또한 원본 확률값(prob)을 독립변수에 실수로 포함시키지 않도록 주의해야 한다는 것과, 정확도(Accuracy)라는 새로운 평가지표를 통해 "전체 예측 중 맞춘 비율"을 계산하는 방법을 익혔습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

오즈비를 구하는 변수가 1단위가 아니라 여러 단위 증가하는 경우로 변형될 수 있습니다.

**변형 예시: 과제제출 건수가 1건이 아니라 3건 증가할 때의 오즈비를 구하라면**

```python
odds_submit_3 = np.exp(model.params['assignment_submit'] * 3)
print("과제제출 3건 증가 시 오즈비:", round(odds_submit_3, 3))
```

`np.exp(계수 * n)`처럼 계수에 증가폭(3)을 곱한 뒤 지수를 취하면 됩니다. 과제제출 건수가 3건 늘어날 때마다 수료 오즈가 약 1.412배 높아진다는 뜻입니다.

| 클래스 불균형 데이터에서 정확도의 한계 | 다수 클래스만 예측해도 정확도가 높게 나올 수 있음 | 민감도·특이도 등 다른 지표와 함께 확인 필요 |
