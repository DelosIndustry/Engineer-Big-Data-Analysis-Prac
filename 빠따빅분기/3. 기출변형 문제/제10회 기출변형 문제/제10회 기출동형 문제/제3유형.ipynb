## 문제 1. (10점)

제공된 데이터(logistics_resignation.csv)는 물류센터 근로자의 인적 특성 및 근무 특성에 관한 자료이다. 나이, 월 급여, 야간근무 횟수 등에 따른 퇴사 여부(resigned)를 예측하기 위해 로지스틱 회귀분석을 수행하고, 아래 소문항에 답하시오.

**제공 데이터 (logistics_resignation.csv)**
| 컬럼명 | 설명 |
|---|---|
| id | 근로자 번호 |
| age | 나이 |
| monthly_wage | 월 급여(만원) |
| night_shift_count | 최근 3개월간 야간근무 횟수 |
| resigned | (종속변수) 퇴사 여부 — 0: 재직, 1: 퇴사 |

**문제 1-1** id를 제외한 모든 변수를 독립변수로 사용하여 로지스틱 회귀모형을 적합하시오. 유의확률(p-value)이 0.05보다 작은 변수의 회귀계수를 구하시오. 단, 절편(상수항)은 제외한다. (반올림하여 소수 셋째 자리까지 작성)

**문제 1-2** 나이(age)가 1 증가할 때 퇴사할 오즈비(Odds Ratio)를 구하시오. (반올림하여 소수 셋째 자리까지 작성)

**문제 1-3** 새로운 근로자(age=35, monthly_wage=260, night_shift_count=12)에 대해, 모델을 이용하여 퇴사 확률을 예측하시오. (반올림하여 소수 셋째 자리까지 작성)

### 문제 풀이 방향

이 문제는 전체 변수로 모형을 적합한 뒤, p-value로 유의한 변수를 찾는 것과 특정 변수(age)의 오즈비를 구하는 것, 그리고 새로운 데이터에 대한 확률 예측까지 이어지는 흐름입니다. 오즈비를 구하는 변수(age)가 반드시 유의한 변수일 필요는 없다는 점이 이번 문제의 특이한 지점입니다. 문제 1-1은 통계적으로 의미 있는 변수를 찾는 질문이고, 문제 1-2는 그 변수의 유의성과 무관하게 "1단위 증가했을 때의 효과"를 해석하는 질문이기 때문입니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

```python
import pandas as pd

df = pd.read_csv('logistics_resignation.csv')
print(df.shape)
df.head()
```

900명의 근로자 데이터입니다. 모든 독립변수가 이미 숫자형이므로 별도의 범주형 인코딩은 필요하지 않습니다.

**셀 2) 로지스틱 회귀모형 적합**

```python
import statsmodels.api as sm

X = df.drop(columns=['id', 'resigned'])
y = df['resigned']
X = sm.add_constant(X)

model = sm.Logit(y, X).fit()
print(model.summary())
```

계수표를 보면 age와 monthly_wage는 p-value가 각각 0.476, 0.417로 전혀 유의하지 않은 반면, night_shift_count는 p-value가 0.000으로 뚜렷하게 유의합니다. 즉 이 데이터 안에서는 나이나 월급보다 야간근무 횟수가 퇴사 여부를 설명하는 데 훨씬 중요한 변수라는 것을 알 수 있습니다.

**셀 3) 유의한 변수 확인**

```python
sig_vars = model.pvalues[model.pvalues < 0.05]
print(sig_vars)
```

`model.pvalues < 0.05`로 조건을 걸면, 절편(const)과 night_shift_count 두 개만 조건을 만족합니다. 문제에서는 절편을 제외하라고 했으므로, 실질적으로 유의한 독립변수는 night_shift_count 하나뿐이라는 것을 확인할 수 있습니다.

**셀 4) 유의한 변수의 회귀계수 (문제 1-1 정답)**

```python
answer_1_1 = round(model.params['night_shift_count'], 3)
print("1-1 정답:", answer_1_1)
```

셀 3에서 절편을 제외하고 유일하게 유의했던 night_shift_count의 계수를 `model.params`에서 바로 꺼냈습니다. 이 문제처럼 유의한 변수가 하나뿐인 경우에는 그 변수의 계수를 바로 답으로 쓰면 되지만, 만약 유의한 변수가 여러 개라면 문제에서 어떤 변수를 지칭하는지 다시 한번 확인해야 합니다.

**셀 5) 나이 1단위 증가 시 오즈비 (문제 1-2 정답)**

```python
import numpy as np

age_odds = np.exp(model.params['age'])
print("1-2 정답:", round(age_odds, 3))
```

age의 계수(-0.0046)에 지수를 취해 오즈비를 구했습니다. age는 셀 3에서 확인했듯 통계적으로 유의하지 않은 변수였지만, "나이가 1살 많아지면 오즈가 몇 배가 되는가"라는 질문 자체는 유의성과 무관하게 계산할 수 있습니다. 오즈비가 0.995로 1에 매우 가깝다는 것은, 나이가 퇴사 오즈에 미치는 영향이 실질적으로 거의 없다는 뜻이며, 이는 셀 2에서 확인한 낮은 유의성과도 일치하는 해석입니다.

> 💡 오즈비 값 자체가 1에 아주 가깝다는 것과, p-value가 커서 통계적으로 유의하지 않다는 것은 같은 현상을 다른 각도에서 보여주는 경우가 많습니다. 오즈비가 1에 가깝다는 것은 "그 변수가 있으나 없으나 결과에 별 차이가 없다"는 뜻이고, 이는 곧 통계적으로도 유의하지 않을 가능성이 높다는 신호입니다.

**셀 6) 새로운 근로자의 퇴사 확률 예측 (문제 1-3 정답)**

```python
new_data = pd.DataFrame({'const': [1], 'age': [35], 'monthly_wage': [260], 'night_shift_count': [12]})
print(new_data)

pred = model.predict(new_data)
print("1-3 정답:", round(pred.values[0], 3))
```

새로운 근로자의 정보를 데이터프레임으로 만들 때, 컬럼 이름과 순서를 모형을 적합할 때 사용한 X와 정확히 맞춰야 합니다. `const` 컬럼도 반드시 포함하고 값은 1로 넣어야 합니다. 이 문제에서는 유의하지 않았던 age, monthly_wage도 그대로 모형에 포함되어 있으므로(셀 4의 문제 1-1과 달리 재적합 과정이 없었습니다), 새 데이터에도 모든 변수를 빠짐없이 넣어야 합니다.

### 전체 코드

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

df = pd.read_csv('logistics_resignation.csv')

X = df.drop(columns=['id', 'resigned'])
y = df['resigned']
X = sm.add_constant(X)

model = sm.Logit(y, X).fit()

# 1-1: 유의한 변수의 회귀계수
answer_1_1 = round(model.params['night_shift_count'], 3)
print("1-1 정답:", answer_1_1)

# 1-2: age 오즈비
age_odds = round(np.exp(model.params['age']), 3)
print("1-2 정답:", age_odds)

# 1-3: 새 데이터 예측
new_data = pd.DataFrame({'const': [1], 'age': [35], 'monthly_wage': [260], 'night_shift_count': [12]})
pred = round(model.predict(new_data).values[0], 3)
print("1-3 정답:", pred)
```

### 정리

이 문제는 재적합 없이 처음 적합한 모형 하나로 세 가지 소문항(유의 계수, 특정 변수 오즈비, 확률 예측)을 모두 처리했습니다. 특히 "유의한 변수의 계수"와 "특정 변수의 오즈비"가 서로 다른 변수(night_shift_count, age)를 가리킬 수 있다는 점에서, 문제가 어떤 변수를 묻고 있는지 각 소문항마다 정확히 확인하는 것이 중요했습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

오즈비 관련 소문항은 "1단위 증가"뿐 아니라 "n단위 증가"로 변형되어 나올 수 있습니다. 이런 변형에는 계수에 증가폭을 곱한 뒤 지수를 취하는 방식으로 대응할 수 있습니다.

**변형 예시: 야간근무 횟수가 1회가 아니라 5회 증가할 때의 오즈비를 구하라면**

```python
odds_5 = np.exp(model.params['night_shift_count'] * 5)
print("야간근무 5회 증가 시 오즈비:", round(odds_5, 3))
```

`np.exp(계수 * 5)`처럼 계수에 증가폭(5)을 곱한 뒤 지수를 취하면 됩니다. night_shift_count는 셀 4에서 확인했듯 강하게 유의한 변수였으므로, 1단위 증가 시 오즈비(약 1.11배)에 비해 5단위 증가 시 오즈비(약 1.68배)가 훨씬 크게 나타납니다. 이는 야간근무가 누적될수록 퇴사 위험이 가파르게 높아진다는 것을 보여줍니다.

## 문제 2. (10점)

제공된 데이터(officetel_rent.csv)는 오피스텔의 매물 특성과 월세에 관한 자료이다. 다중선형 회귀분석을 수행하고 제시된 소문항에 답하시오.

**제공 데이터 (officetel_rent.csv)**
| 컬럼명 | 설명 |
|---|---|
| id | 매물 번호 |
| area | 전용면적(㎡) |
| floor | 층수 |
| station_distance | 최인접 지하철역까지 거리(m) |
| built_year | 준공연도 |
| monthly_rent | (종속변수) 월세(만원) |

**문제 2-1** 모든 독립변수를 포함한 회귀모형을 적합하시오. 이때 절편을 제외한 회귀계수의 합을 구하시오. (반올림하여 소수 셋째 자리까지 작성)

**문제 2-2** 유의한 변수(유의수준 0.05 이하)만을 사용하여 다중회귀분석을 다시 수행하고, R²(결정계수) 값을 구하시오. (반올림하여 소수 셋째 자리까지 작성)

**문제 2-3** 새로운 매물(area=30, station_distance=500, built_year=2015)이 주어졌을 때, 2-2에서 사용한 모델을 이용하여 월세(monthly_rent)를 예측하시오. (반올림하여 소수 셋째 자리까지 작성)

### 문제 풀이 방향

이 문제는 "전체 변수로 적합 → 유의한 변수만 재적합 → 해석 및 예측"이라는 흐름을 다중선형 회귀에 적용합니다. 문제 1과 달리 이번에는 재적합 과정이 포함되어 있으므로, 2-3에서 새로운 매물을 예측할 때는 2-2에서 제외된 변수(floor)를 새 데이터에도 포함시키지 않아야 한다는 점에 유의해야 합니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

```python
import pandas as pd

df = pd.read_csv('officetel_rent.csv')
print(df.shape)
df.head()
```

210개 오피스텔 매물 데이터입니다.

**셀 2) 다중선형 회귀모형 적합**

```python
import statsmodels.api as sm

X = df.drop(columns=['id', 'monthly_rent'])
y = df['monthly_rent']
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
print(model.summary())
```

R-squared가 0.827로, 이 모형이 월세 변동의 약 82.7%를 설명하고 있습니다. 계수표를 보면 area, station_distance, built_year는 p-value가 0.05보다 작아 유의하지만, floor는 p-value가 0.077로 근소하게 유의하지 않습니다. 층수가 높다고 해서 반드시 월세에 뚜렷한 영향을 준다고 보기는 어렵다는 뜻입니다.

**셀 3) 절편 제외 계수 합 (문제 2-1 정답)**


```python
coef_sum = round(model.params.drop('const').sum(), 3)
print("2-1 정답:", coef_sum)
```

`model.params`에서 `.drop('const')`로 절편 행을 제외한 뒤, 남은 네 변수(area, floor, station_distance, built_year)의 계수를 모두 더했습니다.

**셀 4) 유의한 변수 선별**

```python
sig_vars = model.pvalues[model.pvalues < 0.05].index.tolist()
print("유의한 변수(절편 포함):", sig_vars)
```

floor가 제외된 4개(절편 포함)가 유의한 변수로 남았습니다.

**셀 5) 유의한 변수만으로 재적합**

```python
X2 = X[sig_vars]
model2 = sm.OLS(y, X2).fit()
print(model2.summary())
```

floor를 제거하고 재적합한 결과, R-squared는 0.824로 원래 모형(0.827)과 거의 동일하게 유지되어 floor가 애초에 설명력에 별다른 기여를 하지 못했던 변수였다는 것을 다시 확인할 수 있습니다.

**셀 6) R-squared 확인 (문제 2-2 정답)**

```python
r2 = round(model2.rsquared, 3)
print("2-2 정답:", r2)
```

`model.rsquared`는 summary() 표에서 눈으로 확인했던 R-squared 값을 코드로 바로 꺼내는 속성입니다. 재적합한 모형(model2)의 R-squared를 조회했으므로, 셀 5에서 확인한 값과 동일합니다.

**셀 7) 새로운 매물의 월세 예측 (문제 2-3 정답)**

```python
new_data = pd.DataFrame({'const': [1], 'area': [30], 'station_distance': [500], 'built_year': [2015]})
print(new_data)

pred = model2.predict(new_data)
print("2-3 정답:", round(pred.values[0], 3))
```

const를 포함해 2-2 모형에 사용된 컬럼(area, station_distance, built_year)과 정확히 같은 이름과 순서로 새 데이터를 구성했습니다. floor는 2-2 모형에서 이미 제거되었으므로 새 데이터에도 포함하지 않았습니다.

### 전체 코드

```python
import pandas as pd
import statsmodels.api as sm

df = pd.read_csv('officetel_rent.csv')

X = df.drop(columns=['id', 'monthly_rent'])
y = df['monthly_rent']
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

# 2-1: 절편 제외 계수 합
coef_sum = round(model.params.drop('const').sum(), 3)
print("2-1 정답:", coef_sum)

# 유의한 변수만 재적합
sig_vars = model.pvalues[model.pvalues < 0.05].index.tolist()
X2 = X[sig_vars]
model2 = sm.OLS(y, X2).fit()

# 2-2: R-squared
r2 = round(model2.rsquared, 3)
print("2-2 정답:", r2)

# 2-3: 새 데이터 예측
new_data = pd.DataFrame({'const': [1], 'area': [30], 'station_distance': [500], 'built_year': [2015]})
pred = round(model2.predict(new_data).values[0], 3)
print("2-3 정답:", pred)
```

### 정리

문제 2는 문제 1과 달리 재적합 과정을 포함하고 있어서, "절편 포함/제외"라는 조건이 소문항마다 다르게 지시될 수 있다는 것과 재적합 이후에는 제거된 변수를 새 데이터에도 포함시키지 않아야 한다는 것을 함께 다뤘습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

"절편 제외 계수 합"과 "절편 포함 계수 합"은 한 글자 차이(제외/포함)로 완전히 다른 계산이 되는, 이 유형에서 가장 흔한 실수 포인트입니다.

**변형 예시: 절편을 제외하지 않고, 포함한 계수의 합을 구하라면**

```python
coef_sum_with_const = round(model.params.sum(), 3)
print("절편 포함 계수 합:", coef_sum_with_const)
```

`.drop('const')`를 생략하기만 하면 절편까지 포함한 합계가 됩니다. 절편 제외 합(1.806)과 절편 포함 합(-271.912)의 차이가 매우 큰 이유는, 이 데이터의 절편 값(-273.718)이 다른 계수들에 비해 절댓값이 훨씬 크기 때문입니다. 이렇게 절편의 크기가 다른 계수들과 자릿수 차이가 클 때는 "포함/제외"를 헷갈리면 답이 완전히 다른 수준으로 틀어질 수 있으므로, 문제 지문의 표현을 각별히 주의 깊게 읽어야 합니다.

## 이번 회차에서 사용한 핵심 개념 정리
| 상황 | 함수/개념 | 핵심 이유 |
|---|---|---|
| 유의성과 무관하게 특정 변수의 오즈비 구하기 | `np.exp(model.params['변수명'])` | "유의한 변수의 계수"와 "특정 변수의 오즈비"는 다른 질문일 수 있음 |
| 오즈비와 유의성의 관계 | 오즈비가 1에 가까울수록 유의하지 않을 가능성이 높음 | 두 지표가 같은 현상을 다른 각도에서 보여줌 |
| 변수가 n단위 증가할 때의 오즈비 | `np.exp(계수 * n)` | 계수는 1단위 증가당 로그 오즈 증가량 |
| 절편 제외/포함 계수 합 | `.drop('const').sum()` vs `.sum()` | 절편의 크기에 따라 두 결과가 크게 달라질 수 있음 |
| 재적합 이후 새 데이터 구성 | 제거된 변수는 새 데이터에도 포함하지 않음 | 모형이 사용하는 컬럼과 새 데이터의 컬럼이 정확히 일치해야 함 |
