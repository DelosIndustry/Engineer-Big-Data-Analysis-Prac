## 문제 1. (10점)

제공된 데이터(employee_training_test.csv)는 사내 직무교육 테스트에 참여한 직원들의 문제 풀이 결과이다. 각 문제는 대주제, 소주제로 분류되어 있다. 아래 수행 순서에 따라 데이터를 처리한 뒤, 구한 값을 [제출 형식]에 맞춰 답안을 입력하시오.

**[수행 순서]**
1. 소주제별 정답률을 계산한다. (정답률 = 해당 소주제의 정답 개수 / 해당 소주제의 전체 문제 풀이 수)
2. 소주제들의 정답률 중 3번째로 높은 값을 구하시오. 단, 정답률이 같은 경우 하나로 취급한다.

**제공 데이터 (employee_training_test.csv)**
| 컬럼명 | 설명 |
|---|---|
| major_topic | 대주제 |
| sub_topic | 소주제 |
| employee_id | 직원 ID |
| is_correct | 정답 여부 (1: 정답, 0: 오답) |

**[제출 형식]**
- 반올림하여 소수점 셋째 자리까지 작성

### 문제 풀이 방향

한 행이 직원 한 명의 문제 풀이 결과를 나타내므로, 소주제를 기준으로 여러 행을 하나로 묶어서 비율을 구하는 작업이 먼저 필요합니다. 그 다음 정답률 값들 중에서 동률을 하나로 취급한 뒤 3번째로 큰 값을 찾아야 합니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

```python
import pandas as pd

df = pd.read_csv('employee_training_test.csv')
print(df.shape)
print(df.head())
```

565건의 문제 풀이 기록이며, 8개의 소주제(sub_topic)로 나뉘어 있습니다. 대주제(major_topic)는 이번 문제에서는 직접 사용하지 않지만, 데이터 구조를 이해하는 데 참고할 수 있습니다.

**셀 2) 소주제별 정답률 계산하기**

```python
acc = df.groupby('sub_topic')['is_correct'].mean().round(3)
print(acc)
```

`groupby('sub_topic')`은 sub_topic 값이 같은 행들을 하나의 묶음으로 모읍니다. 여기에 `['is_correct'].mean()`을 적용하면 각 묶음 안에서 is_correct 값의 평균이 계산되는데, is_correct는 정답이 1, 오답이 0으로만 되어 있으므로 이 평균이 곧 정답률과 같은 값이 됩니다.

**셀 3) 3번째로 높은 정답률 찾기 (동률 처리 포함, 정답)**

```python
unique_sorted = sorted(set(acc), reverse=True)
answer = unique_sorted[2]

print(unique_sorted)
print("정답:", answer)
```

문제에서 정답률이 같으면 하나로 취급하라고 했으므로, 먼저 중복된 값을 제거해야 합니다. 셀 2의 결과를 자세히 보면 코칭스킬과 피드백이 똑같이 0.75라는 정답률을 가지고 있습니다. `set(acc)`는 파이썬에서 중복을 허용하지 않는 자료형이므로, Series를 집합으로 감싸는 것만으로 중복된 0.75가 자동으로 하나로 합쳐집니다.

여기서 반드시 짚어야 할 함정이 있습니다. 만약 동률을 무시하고 그냥 `acc.sort_values(ascending=False)`로 정렬한 뒤 3번째 행을 그대로 뽑았다면, 정렬 순서상 세 번째 자리에 오는 값은 여전히 0.75(코칭스킬 또는 피드백 중 하나)가 되어버립니다. 즉 "정렬했을 때 3번째 행"과 "중복을 제거한 뒤 3번째로 큰 유일한 값"은 서로 다른 개념이며, 이번 데이터처럼 상위권에 동률이 있는 경우에는 완전히 다른 답이 나옵니다. `sorted(..., reverse=True)`로 큰 값부터 정렬한 뒤, 인덱스 0번이 1등, 1번이 2등, 2번이 3등에 해당하므로 `unique_sorted[2]`가 곧 3번째로 높은 정답률입니다.

> 💡 "동률이면 하나로 취급하라"는 지시가 있는 문제는 정렬만으로 끝내지 말고, 반드시 `set()`이나 `drop_duplicates()`로 중복을 먼저 제거하는 단계를 거쳐야 합니다. 이 단계를 빠뜨리는 것이 이 유형에서 가장 흔한 실수입니다.

### 전체 코드

```python
import pandas as pd

df = pd.read_csv('employee_training_test.csv')

acc = df.groupby('sub_topic')['is_correct'].mean().round(3)
unique_sorted = sorted(set(acc), reverse=True)
answer = unique_sorted[2]

print("정답:", answer)
```

### 정리

이 문제의 핵심은 0과 1로 이루어진 컬럼의 평균이 곧 비율이라는 성질을 이용해 정답률 계산을 한 줄로 끝낸 것과, set()으로 동률을 하나로 합친 뒤 순위를 매긴 것입니다. 동률 처리를 빠뜨리면 답이 실제보다 낮은 순위의 값으로 잘못 나올 수 있으므로, 값이 같으면 하나로 취급하라는 지시가 있을 때는 항상 중복 제거 단계를 먼저 거치는 습관이 필요합니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

기출 복원 자료에서는 "3번째로 높은 값"이라는 조건이 전달되었지만, 실제 시험에서는 몇 번째인지가 다르게 나올 수 있습니다. 이런 유형은 몇 등을 묻는지와 무관하게 "중복 제거 후 인덱스로 조회"라는 동일한 틀로 대응할 수 있어야 합니다.

**변형 예시: 3번째가 아니라 2번째로 높은 값을 구하라면**

```python
answer_2nd = unique_sorted[1]
print("2등(동률 하나로 취급):", answer_2nd)
```

`unique_sorted`는 이미 중복이 제거된 채로 정렬까지 끝난 리스트이므로, 몇 번째를 구하든 인덱스 번호만 바꾸면 됩니다(0번=1등, 1번=2등, 2번=3등). 흥미로운 점은 2등(0.75)이 실제로는 코칭스킬과 피드백 두 소주제가 공유하는 값이라는 것입니다. 즉 "정답률이 2번째로 높은 소주제가 무엇인지"까지 물어보는 문제였다면 정답이 하나가 아니라 두 개(코칭스킬, 피드백)일 수 있다는 점도 함께 기억해두면, 값이 아니라 소주제 이름 자체를 구하라는 변형에도 대비할 수 있습니다.

## 문제 2. (10점)

제공된 데이터(theme_park_sales.csv)는 2024년 3월부터 2025년 8월까지의 테마파크 입장권 판매 이력이다. 소문항을 각각 구한 다음 [제출 형식]에 맞춰 답안을 입력하시오.

**문제 2-1** 연월별 총매출액을 계산하여 큰 순서대로 정렬했을 때, 2번째로 큰 매출액을 입력하시오.

**문제 2-2** 연월별 총매출액이 4번째로 큰 연-월을 찾으시오. 해당 연-월의 데이터에서 입장권 종류(ticket_type)별 매출 합계를 계산하고, 그중 최댓값을 입력하시오.

**제공 데이터 (theme_park_sales.csv)**
| 컬럼명 | 설명 |
|---|---|
| sale_date | 판매 날짜 (YYYY-MM-DD) |
| ticket_type | 입장권 종류 (자유이용권/종일권/야간권/정기권) |
| price | 판매 금액 |

**[제출 형식]**
- 정수로 작성

### 문제 풀이 방향

원본 데이터는 하루 단위(sale_date)로 기록되어 있지만 문제는 월 단위 매출을 요구하므로, 먼저 날짜에서 연-월 정보만 뽑아내는 작업이 필요합니다. 그 다음 2-1은 연월별 매출을 집계해서 등수를 매기는 문제이고, 2-2는 특정 등수의 연월을 먼저 찾은 뒤 그 안에서 다시 입장권 종류 기준으로 한 번 더 집계하는, 두 단계로 이어지는 문제입니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기 + 연월(ym) 컬럼 만들기**

```python
import pandas as pd

df = pd.read_csv('theme_park_sales.csv')
df['ym'] = df['sale_date'].str[:7]

print(df.shape)
print(df.head())
```

`sale_date`는 `'2024-03-01'` 형태의 문자열입니다. `.str[:7]`은 문자열 컬럼 전체에서 앞 7글자만 잘라내는 함수로, 날짜 형식이 `YYYY-MM-DD`로 고정되어 있으므로 앞의 7글자(`2024-03`)만 잘라내면 연-월 정보가 정확히 남습니다.

**셀 2) 연월별 매출 합계를 구하고 정렬하기**

```python
ym_sum = df.groupby('ym')['price'].sum().sort_values(ascending=False)
print(ym_sum)
```

`groupby('ym')['price'].sum()`으로 연월별 매출 합계를 구한 뒤, 바로 `sort_values(ascending=False)`로 큰 값부터 정렬해두었습니다.

**셀 3) 2번째로 큰 매출액 찾기 (문제 2-1 정답)**

```python
answer_2_1 = ym_sum.iloc[1]
print("2-1 정답:", answer_2_1)
```

정렬된 Series에서는 `.iloc[0]`이 1등, `.iloc[1]`이 2등에 해당하므로, 2번째로 큰 매출액은 `.iloc[1]`로 바로 꺼낼 수 있습니다.

**셀 4) 4번째로 큰 연월을 찾고, 그 달의 입장권 종류별 매출 최댓값 구하기 (문제 2-2 정답)**

```python
fourth_ym = ym_sum.index[3]
print("4번째로 큰 연월:", fourth_ym)

sub = df[df['ym'] == fourth_ym]
answer_2_2 = sub.groupby('ticket_type')['price'].sum().sort_values(ascending=False)
print(answer_2_2)
print("2-2 정답:", answer_2_2.max())
```

`ym_sum.index[3]`은 정렬된 Series에서 4번째 행의 인덱스, 즉 연월 이름을 가져옵니다. 그 다음 `df[df['ym'] == fourth_ym]`으로 그 연월에 해당하는 행만 걸러내고, 그 부분 데이터 안에서 다시 `groupby('ticket_type')['price'].sum()`으로 입장권 종류별 합계를 구했습니다.

### 전체 코드

```python
import pandas as pd

df = pd.read_csv('theme_park_sales.csv')
df['ym'] = df['sale_date'].str[:7]

ym_sum = df.groupby('ym')['price'].sum().sort_values(ascending=False)
answer_2_1 = ym_sum.iloc[1]
print("2-1 정답:", answer_2_1)

fourth_ym = ym_sum.index[3]
sub = df[df['ym'] == fourth_ym]
answer_2_2 = sub.groupby('ticket_type')['price'].sum().max()
print("2-2 정답:", answer_2_2)
```

### 정리

이 문제는 날짜에서 필요한 단위(연-월)만 추출하는 전처리, 그룹 집계 후 정렬, 그리고 조건으로 범위를 좁힌 뒤 다시 집계하는 2단계 그룹화까지 한 번에 연습할 수 있는 문제였습니다. 값을 찾는 것(`.iloc`)과 그 값에 해당하는 이름을 찾는 것(`.index`)이 다른 작업이라는 점을 구분하는 것이 중요합니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

이 유형에서 "몇 번째로 큰 값인지"와 "그 범위 안에서 최댓값인지 최솟값인지"는 문제마다 자유롭게 조합될 수 있습니다. 숫자 하나만 바뀌어도 완전히 다른 소문항이 되므로, 정렬과 인덱스 조회 패턴 자체를 이해해두는 것이 중요합니다.

**변형 예시: 5번째로 큰 연월을 찾아, 그 안에서 최솟값을 구하라면**

```python
fifth_ym = ym_sum.index[4]
print("5번째로 큰 연월:", fifth_ym, "/ 매출액:", ym_sum.iloc[4])

sub5 = df[df['ym'] == fifth_ym]
cat_sum5 = sub5.groupby('ticket_type')['price'].sum().sort_values(ascending=False)
print(cat_sum5)
print("최솟값:", cat_sum5.min())
```

바뀐 부분은 인덱스 번호(3 → 4)와 `.max()`가 `.min()`으로 바뀐 것뿐입니다. "몇 번째"라는 순번과 "최댓값/최솟값"이라는 방향, 이 두 가지 축을 각각 독립적으로 바꿀 수 있다는 것을 이해해두면, 두 조건이 어떻게 조합되어도 같은 코드 구조로 대응할 수 있습니다.

## 문제 3. (10점)

제공된 데이터(comment_data.csv)는 온라인 게시판 댓글의 내용과 정상/악성 여부를 담고 있다. 아래 수행 순서에 따라 데이터를 처리한 뒤, 구한 값을 [제출 형식]에 맞춰 답안을 입력하시오.

**[수행 순서]**
1. 각 댓글의 단어 개수를 띄어쓰기(공백) 기준으로 계산하시오.
2. 악성 댓글의 평균 단어 개수와 정상 댓글의 평균 단어 개수를 각각 구하시오.
3. 두 평균의 차이의 절댓값을 구하시오.

**제공 데이터 (comment_data.csv)**
| 컬럼명 | 설명 |
|---|---|
| label | 댓글 구분 (정상/악성) |
| text | 댓글 내용 |

**[제출 형식]**
- 반올림하여 소수점 셋째 자리까지 작성

### 문제 풀이 방향

이 문제는 텍스트라는 재료를 다루지만, 결국 핵심은 텍스트를 숫자(단어 개수)로 바꾸는 전처리 한 단계를 거친 뒤에는 그룹별 평균을 비교하는 익숙한 흐름으로 이어진다는 것입니다.


### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

```python
import pandas as pd

df = pd.read_csv('comment_data.csv')
print(df.shape)
print(df.head())
```

320개의 댓글 데이터이며, label이 '정상'과 '악성' 두 가지로 나뉘어 있습니다.

**셀 2) 공백 기준으로 단어 개수 세기**

```python
df['words'] = df['text'].apply(lambda x: len(x.split(' ')))
print(df[['label', 'text', 'words']].head())
```

`x.split(' ')`은 문자열 x를 공백 기준으로 잘라서 리스트로 만들고, 그 리스트의 길이가 곧 단어 개수가 됩니다. `.apply(lambda x: ...)`는 컬럼의 각 값에 대해 이 동작을 반복 적용해주는 함수입니다.

**셀 3) label별 평균 단어 개수 구하기**

```python
group_mean = df.groupby('label')['words'].mean()
print(group_mean)
```

`groupby('label')`로 정상과 악성을 각각 하나의 그룹으로 나누고, `['words'].mean()`으로 그룹 안에서 단어 개수의 평균을 구했습니다. 악성 댓글이 평균적으로 정상 댓글보다 단어 수가 더 많다는 것을 확인할 수 있습니다.

**셀 4) 두 평균의 차이(절댓값) 구하기 (정답)**

```python
answer = round(abs(group_mean['정상'] - group_mean['악성']), 3)
print("정답:", answer)
```

`group_mean`은 label을 인덱스로 가지는 Series이므로, `group_mean['정상']`과 `group_mean['악성']`처럼 인덱스 이름으로 각 그룹의 평균값을 바로 꺼낼 수 있습니다. 두 값을 뺀 뒤 `abs()`로 절댓값을 취해서, 어느 쪽이 더 큰지와 무관하게 차이의 크기만 구했습니다.

### 전체 코드

```python
import pandas as pd

df = pd.read_csv('comment_data.csv')
df['words'] = df['text'].apply(lambda x: len(x.split(' ')))

group_mean = df.groupby('label')['words'].mean()
answer = round(abs(group_mean['정상'] - group_mean['악성']), 3)

print("정답:", answer)
```

### 정리

이 문제는 텍스트를 숫자(단어 개수)로 바꾸는 전처리 한 단계를 거친 뒤에는 지금까지 계속 봐온 groupby().mean() 패턴으로 그대로 이어진다는 것을 보여줍니다. 데이터의 종류가 숫자든 텍스트든, 일단 비교하고 싶은 대상을 숫자 컬럼으로 만들어두면 그 다음 분석 흐름은 동일하게 적용된다는 점을 기억해두면 좋습니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

이 유형은 "평균"이 아니라 다른 통계량(중앙값, 최댓값 등)의 차이를 물어보는 방식으로 변형될 수 있습니다. 또한 "공백 기준 단어 수"가 아니라 "글자 수"(공백 포함/제외)를 기준으로 물어볼 수도 있습니다.

**변형 예시: 평균이 아니라 각 그룹의 최대 단어 수 차이를 구하라면**

```python
group_max = df.groupby('label')['words'].max()
print(group_max)

answer_max = abs(group_max['정상'] - group_max['악성'])
print("최댓값 기준 차이:", answer_max)
```

`.mean()`을 `.max()`로 바꾸기만 하면 됩니다. 평균 기준 차이(2.5)와 최댓값 기준 차이(2)가 서로 다른 숫자로 나온다는 점에 주목할 필요가 있습니다. "평균적으로 얼마나 차이 나는가"와 "가장 긴 댓글끼리 비교하면 얼마나 차이 나는가"는 서로 다른 질문이므로, 문제 지문에서 정확히 어떤 통계량을 요구하는지(평균/중앙값/최댓값/최솟값)를 놓치지 않고 읽는 것이 중요합니다.

## 이번 회차에서 사용한 핵심 함수 정리
| 상황 | 함수 | 핵심 이유 |
|---|---|---|
| 0/1 값으로 비율(정답률) 구하기 | `groupby().mean()` | 이진 값의 평균이 곧 비율이 된다는 성질 이용 |
| 값이 같으면 하나로 취급 (동률 처리) | `set()` | 중복을 허용하지 않는 자료형이라 자동으로 중복 제거 |
| 날짜 문자열에서 연-월만 추출 | `.str[:7]` | 고정된 날짜 형식에서 필요한 구간만 슬라이싱 |
| 그룹 합계 후 등수 조회 | `groupby().sum().sort_values()` + `.iloc[n-1]` | 정렬 후 인덱스 위치로 등수를 바로 조회 |
| 값이 아니라 이름(인덱스)이 필요할 때 | `.index[n]` | iloc은 값의 위치, index는 그 위치의 이름 |
| 텍스트를 숫자로 변환하기 (단어 개수 등) | `.apply(lambda x: ...)` | 내장 함수로 처리되지 않는 개별적인 로직을 컬럼 전체에 적용 |
| 그룹별 통계량 비교 (평균/최댓값 등) | `groupby()[대상].mean()` / `.max()` 등 | 요구하는 통계량에 따라 함수만 바꿔서 재사용 |
