## 문제 1. (10점)

제공된 데이터(solar_capacity.csv)는 2003년부터 2023년까지 20개 국가의 연도별 태양광 발전설비 누적 용량(GW) 데이터입니다. 아래 소문제를 각각 구하시오.

**문제 1-1** 2003년부터 2023년까지 각 연도별로 태양광 발전설비 용량이 가장 큰 국가를 찾고, 그 중 1위를 가장 많이 차지한 국가의 1위 횟수를 정수로 구하시오.

**문제 1-2** 다음 공식으로 각 국가의 CAGR(연평균 성장률)을 계산하고, 그 중 최댓값을 소수점 셋째 자리까지 구하시오.

```
CAGR = (EV / BV)^(1/n) - 1
```
- EV(Ending Value): 최종 값 (2023년 용량)
- BV(Beginning Value): 초기 값 (2003년 용량)
- n: 전체 기간 (2023년 - 2003년 = 20년)

**제공 데이터 (solar_capacity.csv)**
| 컬럼명 | 설명 |
|---|---|
| country | 국가명 (20개국) |
| 2003 ~ 2023 | 해당 연도의 태양광 발전설비 누적 용량(GW) |

**[제출 형식]**
- 문항 1-1: 정수(integer)로 작성
- 문항 1-2: 반올림하여 소수점 셋째 자리까지 작성

### 문제 풀이 방향

이 데이터는 국가가 행, 연도가 열로 되어 있는 가로형(wide format) 데이터입니다. 1-1은 각 열(연도)에서 값이 가장 큰 행(국가)을 21번 반복해서 찾은 뒤 그 결과를 집계하는 문제이고, 1-2는 국가별로 시작값과 끝값만 있으면 계산되는 공식이므로 컬럼 전체를 한 번에 계산하는 방식으로 풀 수 있습니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

````python
import pandas as pd

df = pd.read_csv('solar_capacity.csv')
print(df.shape)
df[['country', '2003', '2004', '2023']].head()
````

20개국, 21개 연도(2003~2023)에 대한 용량 데이터입니다. 컬럼명이 "2003"처럼 숫자로 보이지만 실제로는 문자열이므로, 뒤에서 컬럼을 조회할 때는 `df['2003']`처럼 문자열로 접근해야 합니다.

**셀 2) 연도별 1위 국가 찾기**

````python
year_cols = [str(y) for y in range(2003, 2024)]
top_country_by_year = df.set_index('country')[year_cols].idxmax(axis=0)
print(top_country_by_year.head())
````

`set_index('country')`로 국가명을 인덱스로 지정해두면, `idxmax(axis=0)`의 결과가 행 번호가 아니라 국가 이름으로 바로 나옵니다. `axis=0`은 각 컬럼(연도)을 위에서 아래 방향으로 훑으며 최댓값의 위치를 찾는다는 뜻입니다.

**셀 3) 최다 1위 국가의 횟수 구하기 (문제 1-1 정답)**

````python
count_result = top_country_by_year.value_counts()
answer_1_1 = count_result.max()

print(count_result)
print("문제 1-1 정답:", answer_1_1)
````

`value_counts()`로 21개 연도의 1위 국가 목록에서 각 국가가 몇 번 등장했는지 셌습니다. 초반 몇 년은 칠레·이탈리아·멕시코가 1위였다가, 중국이 빠르게 성장하면서 이후 대부분의 해에 1위를 차지한 것으로 보입니다.

**셀 4) 국가별 CAGR 계산하기**

````python
BV, EV, n = df['2003'], df['2023'], 20
df['CAGR'] = (EV / BV) ** (1/n) - 1

print(df[['country', 'CAGR']].head())
````

`n = 20`인 이유는 2003년부터 2023년까지 연도는 21개(2003, ..., 2023)이지만 성장이 일어나는 구간은 2003→2004, ..., 2022→2023까지 20번이기 때문입니다.

**셀 5) CAGR 최댓값 찾기 (문제 1-2 정답)**

````python
answer_1_2 = round(df['CAGR'].max(), 3)
print("문제 1-2 정답:", answer_1_2)
````

CAGR 컬럼 전체에서 최댓값을 찾아 반올림했습니다. 셀 3의 idxmax()와 마찬가지로, 이번에는 값 자체(어느 국가인지가 아니라 CAGR 수치)가 필요하므로 idxmax가 아니라 max()를 사용했습니다.

### 전체 코드

````python
import pandas as pd

df = pd.read_csv('solar_capacity.csv')
year_cols = [str(y) for y in range(2003, 2024)]

# 1-1
top_country_by_year = df.set_index('country')[year_cols].idxmax(axis=0)
answer_1_1 = top_country_by_year.value_counts().max()
print("문제 1-1 정답:", answer_1_1)

# 1-2
BV, EV, n = df['2003'], df['2023'], 20
df['CAGR'] = (EV / BV) ** (1/n) - 1
answer_1_2 = round(df['CAGR'].max(), 3)
print("문제 1-2 정답:", answer_1_2)
````

### 정리

이 문제는 지금까지 여러 회차에서 반복적으로 등장한 "연도별 1위 최다 국가 + CAGR" 패턴을 새로운 도메인(태양광 발전 용량)으로 재구성한 것입니다. idxmax로 연도별 위치를 찾고, 컬럼 간 벡터 연산으로 CAGR을 한 번에 계산한다는 핵심 흐름은 동일합니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

CAGR 최댓값이 아니라 최솟값(가장 느리게 성장한 국가)을 물어보는 방식으로 변형될 수 있습니다.

**변형 예시: 최댓값이 아니라 최솟값(가장 느리게 성장한 국가)을 구하라면**

````python
min_cagr = round(df['CAGR'].min(), 3)
min_country = df.loc[df['CAGR'].idxmin(), 'country']
print("최소 CAGR:", min_cagr, "/ 국가:", min_country)
````

`.max()`가 `.min()`으로, `idxmax()`가 `idxmin()`으로 바뀌기만 하면 됩니다. 이 데이터에서는 모든 국가의 CAGR이 양수(성장)인데, 실제 시험 데이터에서는 특정 국가가 음수 CAGR(용량 감소)을 보일 수도 있습니다. 만약 최솟값이 음수로 나온다면, 그 국가는 해당 기간 동안 오히려 용량이 줄어들었다는 의미로 해석해야 합니다.

## 문제 2. (10점)

제공된 데이터(smartfarm_sensor.csv)는 스마트팜에서 시간 단위로 수집한 센서 측정값입니다. 아래 수행 순서에 따라 데이터를 처리한 뒤, 구한 값을 [제출 형식]에 맞춰 답안을 입력하시오.

**[수행 순서]**
1. 결측치가 가장 많은 컬럼을 찾으시오.
2. 해당 컬럼의 결측치를 중앙값(median)으로 대체하시오.
3. 결측치를 대체한 후 해당 컬럼의 평균을 구하시오.

**제공 데이터 (smartfarm_sensor.csv)**
| 컬럼명 | 설명 |
|---|---|
| DateTime | 측정 일시 |
| Temp | 온도(℃) |
| Humidity | 습도(%) |
| SoilMoisture | 토양 수분(%) |
| CO2 | 이산화탄소 농도(ppm) |
| Light | 조도(lux) |

**[제출 형식]**
- 반올림하여 소수점 셋째 자리까지 작성

### 문제 풀이 방향

지금까지 다룬 결측치 문제들은 "어느 컬럼에 결측치가 있는지"를 문제에서 미리 알려줬지만, 이번 문제는 **어느 컬럼이 결측치가 가장 많은지를 직접 코드로 찾아내는 것부터 시작**합니다. 여러 컬럼의 결측치 개수를 한 번에 세고, 그 중 최댓값을 가진 컬럼의 이름을 알아낸 뒤에야 비로소 이후 단계(대체, 평균 계산)를 진행할 수 있습니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

````python
import pandas as pd

df = pd.read_csv('smartfarm_sensor.csv')
print(df.shape)
print(df.head())
````

500개의 시간별 측정 기록이며, 여러 센서 컬럼에 NaN이 섞여 있는 것을 첫 5행만 봐도 확인할 수 있습니다.

**셀 2) 컬럼별 결측치 개수 확인하기**

````python
missing_counts = df.isna().sum()
print(missing_counts)
````

`df.isna()`는 데이터프레임의 모든 칸을 결측치 여부(True/False)로 바꾼 표를 만들고, 여기에 `.sum()`을 붙이면 컬럼별로 True(결측치)의 개수를 세어줍니다. DateTime에는 결측치가 없고, 나머지 다섯 개 센서 컬럼에는 각각 다른 개수의 결측치가 있습니다.

**셀 3) 결측치가 가장 많은 컬럼 찾기**

````python
max_col = missing_counts.idxmax()
print("결측치가 가장 많은 컬럼:", max_col, "/ 개수:", missing_counts.max())
````

`missing_counts`는 컬럼명을 인덱스로, 결측치 개수를 값으로 가지는 Series입니다. 여기에 `idxmax()`를 적용하면 값이 가장 큰(결측치가 가장 많은) 인덱스, 즉 컬럼 이름이 바로 나옵니다. 이 셀이 이번 문제의 핵심으로, 컬럼명을 눈으로 찾지 않고 코드로 자동 탐지했다는 점이 중요합니다. 이렇게 자동으로 찾아두면, 실제 시험에서 어떤 컬럼에 결측치가 가장 많은지 매번 데이터가 달라져도 똑같은 코드로 대응할 수 있습니다.

**셀 4) 해당 컬럼의 결측치를 중앙값으로 대체하기**

````python
median_val = df[max_col].median()
df[max_col] = df[max_col].fillna(median_val)

print("대체한 중앙값:", median_val)
print(df[max_col].isna().sum())
````

`df[max_col]`처럼 변수에 담긴 컬럼명으로 실제 컬럼에 접근할 수 있습니다. 컬럼명을 문자열로 직접 타이핑하지 않고 `max_col`이라는 변수를 사용했기 때문에, 셀 3에서 찾은 컬럼이 무엇이든(Humidity가 아니라 다른 컬럼이었더라도) 이 코드는 그대로 재사용됩니다.

**셀 5) 대체 후 평균 계산 (정답)**

````python
answer = round(df[max_col].mean(), 3)
print("정답:", answer)
````

중앙값으로 결측치를 채운 뒤 다시 그 컬럼의 평균을 구했습니다. 중앙값(60.365)과 대체 후 평균(60.254)이 비슷하면서도 정확히 같지는 않은 값으로 나온 것을 확인할 수 있는데, 이는 원래 있던 값들의 분포와 새로 채워 넣은 중앙값이 뒤섞이면서 평균이 살짝 달라지기 때문입니다.

### 전체 코드

````python
import pandas as pd

df = pd.read_csv('smartfarm_sensor.csv')

# 결측치가 가장 많은 컬럼 찾기
missing_counts = df.isna().sum()
max_col = missing_counts.idxmax()

# 해당 컬럼의 결측치를 중앙값으로 대체
median_val = df[max_col].median()
df[max_col] = df[max_col].fillna(median_val)

# 대체 후 평균
answer = round(df[max_col].mean(), 3)
print("결측치 최다 컬럼:", max_col)
print("정답:", answer)
````

### 정리

이 문제의 핵심은 `isna().sum()`으로 컬럼별 결측치 개수를 세고, `idxmax()`로 그 중 가장 큰 값을 가진 컬럼 이름을 자동으로 찾아냈다는 것입니다. 이후 컬럼명을 문자열로 직접 쓰지 않고 변수(`max_col`)로 참조했기 때문에, 앞 단계에서 찾은 결과가 뒤 단계까지 자연스럽게 이어집니다. "어떤 컬럼인지 직접 찾아야 하는" 문제는 이렇게 컬럼명을 변수에 담아 재사용하는 습관이 특히 중요합니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

"가장 많은" 컬럼이 아니라 "두 번째로 많은" 컬럼을 물어보거나, 특정 컬럼을 제외하고 찾으라는 조건이 추가될 수 있습니다.

**변형 예시: 결측치가 두 번째로 많은 컬럼을 구하라면**

````python
missing_counts_no_date = missing_counts.drop('DateTime')
sorted_counts = missing_counts_no_date.sort_values(ascending=False)
print(sorted_counts)

second_col = sorted_counts.index[1]
print("두 번째로 결측치가 많은 컬럼:", second_col, "/ 개수:", sorted_counts.iloc[1])
````

`idxmax()`는 1등만 찾아줄 뿐 2등, 3등은 알려주지 않으므로, `sort_values(ascending=False)`로 정렬한 뒤 `.index[1]`로 두 번째 행의 컬럼명을 조회하는 방식으로 바꿨습니다. 이때 결측치가 애초에 없는 DateTime처럼, 애초에 결측치를 셀 필요가 없는 컬럼(날짜, ID 등)은 `.drop()`으로 미리 제외해두면 등수를 셀 때 불필요한 혼란을 막을 수 있습니다.

## 문제 3. (10점)

제공된 데이터(travel_booking.csv)는 여행 패키지 예약 데이터입니다. 전처리 후 아래 소문제를 각각 구하시오.

**[전처리 수행]**
- 새로운 컬럼 `order_amt` 생성: `order_amt = people_count × unit_price`
- 새로운 컬럼 `cancel_TF` 생성: 예약번호(`booking_no`)가 'C'로 시작하면 True, 아니면 False ('C'는 취소 예약)
- `cancel_TF`가 True인 행의 `order_amt`를 음수로 변환

**문제 3-1** 취소 예약 중 `order_amt`의 절댓값이 가장 큰 예약번호를 구하시오.

**문제 3-2** 고객ID(`customer_id`)별로 `order_amt`를 합산하여 가장 큰 값을 소수점 둘째 자리까지 구하시오.

**제공 데이터 (travel_booking.csv)**
| 컬럼명 | 설명 |
|---|---|
| booking_no | 예약번호 |
| customer_id | 고객 ID |
| package_name | 여행 패키지명 |
| people_count | 예약 인원 수 |
| unit_price | 1인당 가격(원) |

**[제출 형식]**
- 문항 3-1: 문자열(예약번호) 그대로 작성
- 문항 3-2: 반올림하여 소수점 둘째 자리까지 작성

### 문제 풀이 방향

전처리 지시사항을 순서대로 적용한 뒤, 취소 예약만 따로 떼어내서 크기를 비교하는 문제(3-1)와 고객이라는 기준으로 묶어서 합계를 비교하는 문제(3-2)를 차례로 풉니다.

### 정답 코드 및 해설

**셀 1) 데이터 불러오기**

````python
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

df = pd.read_csv('travel_booking.csv')
print(df.shape)
print(df.head())
````

정상 예약은 'B'로, 취소 예약은 'C'로 시작하도록 예약번호가 설계되어 있습니다.

**셀 2) order_amt, cancel_TF 생성 및 부호 변환**

````python
df['order_amt'] = df['people_count'] * df['unit_price']
df['cancel_TF'] = df['booking_no'].str.startswith('C')
df.loc[df['cancel_TF'], 'order_amt'] = -df.loc[df['cancel_TF'], 'order_amt']

print(df.head())
````

인원수와 단가를 곱해 예약금액을 만들고, 예약번호 첫 글자로 취소 여부를 판별한 뒤, 취소 건의 금액에 마이너스 부호를 붙였습니다.

**셀 3) 취소 예약 중 절댓값이 가장 큰 예약번호 (문제 3-1 정답)**

````python
cancel_df = df[df['cancel_TF']]
answer_3_1 = cancel_df.loc[cancel_df['order_amt'].abs().idxmax(), 'booking_no']

print("문제 3-1 정답:", answer_3_1)
````

취소 예약만 걸러낸 뒤(`cancel_df`), 절댓값이 가장 큰 행의 위치를 `idxmax()`로 찾아 그 예약번호를 조회했습니다. `order_amt`는 취소 건이라면 이미 음수이므로, 부호를 무시한 크기(절댓값)로 비교해야 취소 규모가 가장 큰 예약을 정확히 찾을 수 있습니다.

**셀 4) 고객별 합산 최댓값 (문제 3-2 정답)**

````python
cus_sum = df.groupby('customer_id')['order_amt'].sum()
answer_3_2 = round(cus_sum.max(), 2)

print(cus_sum.sort_values(ascending=False).head())
print("문제 3-2 정답:", answer_3_2)
````

`groupby('customer_id')['order_amt'].sum()`으로 고객별 합계를 구했습니다. 취소 건은 이미 음수로 처리되어 있으므로, 이 합계에는 정상 예약(+)과 취소 예약(-)이 자연스럽게 섞여 계산되어 고객별 실질 결제금액이 나옵니다.

### 전체 코드

````python
import pandas as pd

df = pd.read_csv('travel_booking.csv')

df['order_amt'] = df['people_count'] * df['unit_price']
df['cancel_TF'] = df['booking_no'].str.startswith('C')
df.loc[df['cancel_TF'], 'order_amt'] = -df.loc[df['cancel_TF'], 'order_amt']

# 3-1
cancel_df = df[df['cancel_TF']]
answer_3_1 = cancel_df.loc[cancel_df['order_amt'].abs().idxmax(), 'booking_no']
print("문제 3-1 정답:", answer_3_1)

# 3-2
answer_3_2 = round(df.groupby('customer_id')['order_amt'].sum().max(), 2)
print("문제 3-2 정답:", answer_3_2)
````

### 정리

이 문제는 부호로 취소 여부를 표현해두면 이후 집계가 조건문 없이 단순해진다는 원칙을 다시 확인했습니다. 절댓값으로 크기를 비교할 때(3-1)와 부호를 그대로 살려서 합산할 때(3-2)를 구분해야 한다는 점도 동일하게 적용됩니다.

### 🔍 실전 대비: 예상되는 변형과 접근 전략

고객별 합산 "최댓값"이 아니라 "최솟값"을 물어보면, 취소가 유난히 많았던(순매출이 마이너스인) 고객을 찾는 문제가 됩니다.

**변형 예시: 고객별 합산 금액이 가장 작은(가장 손실이 큰) 고객을 구하라면**

````python
cus_sum_sorted = cus_sum.sort_values()
print(cus_sum_sorted.head())

min_customer = cus_sum_sorted.index[0]
min_value = cus_sum_sorted.iloc[0]
print("합산 금액이 가장 작은 고객:", min_customer, "/ 금액:", min_value)
````

`sort_values()`를 오름차순(기본값) 그대로 사용하면 가장 작은 값이 맨 위로 오므로, `.iloc[0]`으로 최솟값을, `.index[0]`으로 그 고객 ID를 바로 조회할 수 있습니다. 합산 금액이 음수라는 것은, 이 고객은 정상 예약보다 취소 예약의 금액 합이 더 컸다는 뜻입니다.

## 이번 회차에서 사용한 핵심 함수 정리
| 상황 | 함수 | 핵심 이유 |
|---|---|---|
| 컬럼별 결측치 개수 한 번에 세기 | `df.isna().sum()` | 모든 컬럼을 True/False로 바꾼 뒤 컬럼별 합계 |
| 결측치가 가장 많은 컬럼 이름 찾기 | `missing_counts.idxmax()` | 값이 아니라 그 값을 가진 인덱스(컬럼명)가 필요할 때 |
| 찾아낸 컬럼명을 이후 코드에서 재사용 | `df[변수명]`처럼 변수에 담아 참조 | 컬럼명을 문자열로 직접 쓰지 않아 다른 데이터에도 코드 재사용 가능 |
| 컬럼(연도)별 최댓값의 위치(국가) 찾기 | `idxmax(axis=0)` | 값이 아니라 값의 위치(행)가 필요할 때 |
| 문자열 조건 확인 | `str.startswith('X')` | 문자열 함수를 컬럼 전체에 한 번에 적용 |
| 조건에 맞는 값만 변경 | `df.loc[조건, 컬럼] = 값` | 조건을 만족하는 행만 정확히 선택하여 값 변경 |
| 취소 등 부호가 섞인 데이터의 합계 | `groupby().sum()` | 부호를 미리 반영해두면 조건 분기 없이 합산 가능 |
