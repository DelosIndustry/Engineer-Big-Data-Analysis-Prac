## 공통 준비

```python
import pandas as pd

df = pd.read_csv('data/part01_pandas/pandas_preprocess.csv')
print(df.shape)
print(df.head())
```

## 문제 01

spend 컬럼의 IQR 기준 상한값을 초과하는 값을 상한값으로 대체했을 때, MB0002의 spend 값을 구하시오.

**정답 코드**

```python
q1 = df['spend'].quantile(0.25)
q3 = df['spend'].quantile(0.75)
iqr = q3 - q1
upper = q3 + 1.5 * iqr

df_capped = df.copy()
df_capped.loc[df_capped['spend'] > upper, 'spend'] = upper

answer = df_capped.loc[
    df_capped['member_id'] == 'MB0002', 'spend'
].iloc[0]
print('upper:', upper)
print('answer:', answer)
```

**간단 해설**

MB0002의 원래 spend는 980000.0이지만, IQR 기준 상한값 462000.0을 초과한다. 상한값으로 대체하면 MB0002의 spend는 462000.0이 된다.

## 문제 02

signup_date 컬럼을 날짜형으로 변환한 뒤, 가입 연도가 2026년인 행의 개수를 구하시오.

**정답 코드**

```python
df['signup_date'] = pd.to_datetime(df['signup_date'])
answer = (df['signup_date'].dt.year == 2026).sum()
print(answer)
```

**간단 해설**

signup_date를 날짜형으로 바꾸면 dt.year로 연도를 추출할 수 있다. 추출한 연도가 2026인 행을 True로 표시하고 합계를 구하면 28개이다.

---

**정답** 1. 462000.0, 2. 28
