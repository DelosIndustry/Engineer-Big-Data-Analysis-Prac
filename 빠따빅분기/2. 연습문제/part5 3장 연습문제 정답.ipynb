## 공통 준비

```python
import pandas as pd
import statsmodels.api as sm

orders_path = (
    'data/part04_type3/correlation_regression_ad_orders/'
    'ad_orders.csv'
)
orders = pd.read_csv(orders_path)
print(orders.shape)
print(orders.head())
```

## 문제 01

ad_budget과 is_weekend를 사용한 다중선형 회귀모형을 적합하시오. 주말 여부를 1로 고정하고 광고비가 90인 경우와 130인 경우의 주문 수를 예측한 뒤 두 예측값의 차이를 구하시오.

**정답 코드**

```python
multi_X = sm.add_constant(
    orders[['ad_budget', 'is_weekend']]
)
multi_model = sm.OLS(orders['orders'], multi_X).fit()

new_data = pd.DataFrame({
    'const': [1, 1],
    'ad_budget': [90, 130],
    'is_weekend': [1, 1]
})
predicted = multi_model.predict(new_data)

print('pred_90:', format(predicted.iloc[0], '.4f'))
print('pred_130:', format(predicted.iloc[1], '.4f'))
print(
    'difference:',
    format(predicted.iloc[1] - predicted.iloc[0], '.4f')
)
```

**간단 해설**

주말 여부를 같게 두었으므로 두 예측값의 차이는 광고비가 40 증가한 경우의 모형 예측 차이다. 광고비 90과 130의 예측 주문 수는 각각 약 283.4924와 364.1425이며 차이는 80.6500이다.
