## 문제 01

sales 데이터에서 region을 행, category를 열로 하는 순매출 합계 표를 만드시오. 각 region에서 합계가 가장 큰 category를 찾은 뒤, 가장 자주 1위를 기록한 category와 횟수를 모두 구하시오.

**정답 코드**

```python
pivot = sales.pivot_table(
    index='region',
    columns='category',
    values='net_sales',
    aggfunc='sum'
)
top_category = pivot.idxmax(axis=1)
counts = top_category.value_counts()

max_count = counts.max()
top_names = counts[counts == max_count].index.tolist()

print(pivot.round(1))
print(top_category)
print('top_categories:', top_names)
print('count:', int(max_count))
```

**간단 해설**

idxmax(axis=1)은 각 지역의 행에서 가장 큰 카테고리 열을 반환한다. 빈도를 구한 뒤 최댓값과 같은 항목을 모두 선택하면 동률도 빠뜨리지 않는다.

## 문제 02

status가 방문인 행만 보았을 때 방문 건수가 가장 많은 service_group과 건수를 구하시오.

**정답 코드**

```python
reservations['service_group'] = (
    reservations['service_code'].str.split('-').str[0]
)
visit_counts = reservations[
    reservations['status'] == '방문'
]['service_group'].value_counts()

print(visit_counts)
print('top_visit_group:', visit_counts.idxmax())
print('top_visit_count:', int(visit_counts.max()))
```

**간단 해설**

방문 행을 먼저 필터링한 뒤 service_group의 value_counts()를 구한다. idxmax() 또는 index[0]으로 그룹명, max() 또는 iloc[0]으로 건수를 구한다.
