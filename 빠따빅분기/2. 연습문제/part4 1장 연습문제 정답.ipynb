## 공통 준비

```python
import pandas as pd
from scipy import stats

path = 'data/part04_type3/one_sample_ttest_study_time/study_time.csv'
df = pd.read_csv(path)
x = df['study_hours']
print(df.shape)
print(df.head())
```

## 문제 01

이번 데이터의 95% 신뢰구간 하한과 상한을 구하고, 기준값 40의 포함 여부를 쓰시오.

**정답 코드**

```python
result = stats.ttest_1samp(x, popmean=40)
ci = result.confidence_interval(confidence_level=0.95)

print('ci_lower:', round(ci.low, 4))
print('ci_upper:', round(ci.high, 4))
print('contains_40:', ci.low <= 40 <= ci.high)
```

**간단 해설**

95% 신뢰구간은 [41.8641, 44.9415]이며 40은 하한보다 작아 구간에 포함되지 않는다.

## 문제 02

유의수준 0.05에서 p-value가 0.000074이고 대립가설이 '모평균은 40과 다르다'일 때 귀무가설 판단과 결론을 쓰시오.

**정답 코드**

```python
popmean = 40
alpha = 0.05
result = stats.ttest_1samp(x, popmean=popmean, alternative='two-sided')

if result.pvalue < alpha:
    decision = '귀무가설 기각'
    interpretation = '평균 학습 시간은 40시간과 통계적으로 유의한 차이가 있다.'
else:
    decision = '귀무가설 기각하지 못함'
    interpretation = '평균 학습 시간이 40시간과 다르다고 볼 근거가 부족하다.'

print('p_value:', format(result.pvalue, '.6f'))
print('decision:', decision)
print('interpretation:', interpretation)
```

**간단 해설**

p-value가 0.05보다 작으므로 귀무가설을 기각한다. 대립가설이 '다르다'이므로 기준값과 통계적으로 유의한 차이가 있다고 해석한다.

---

**정답** 1. 하한 41.8641, 상한 44.9415, 40은 포함되지 않음, 2. 귀무가설 기각, 평균 학습 시간은 40시간과 통계적으로 유의한 차이가 있다.
