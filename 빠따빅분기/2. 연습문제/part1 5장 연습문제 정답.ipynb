## 공통 준비

```python
import pandas as pd

orders = pd.read_csv('data/part01_pandas/pandas_orders.csv')
print(orders.shape)
print(orders.head())
```

## 문제 01

category별 sales_amount 합계를 구한 뒤 매출 합계가 가장 큰 상품 분류명을 출력하시오.

**정답 코드**

```python
category_sales = orders.groupby('category')['sales_amount'].sum()
print(category_sales.idxmax())
```

**간단 해설**

category별 매출 합계를 만든 뒤 idxmax()를 사용하면 가장 큰 값의 인덱스명을 구할 수 있다.

## 문제 02

qty, unit_price, discount_rate 중 sales_amount와의 상관계수 절댓값이 가장 큰 컬럼명을 구하시오.

**정답 코드**

```python
num = orders[['qty', 'unit_price', 'discount_rate', 'sales_amount']]
corr = num.corr()
answer = corr['sales_amount'].drop('sales_amount').abs().idxmax()
print(answer)
```

**간단 해설**

sales_amount와 자기 자신의 값은 제외하고 절댓값이 가장 큰 인덱스를 idxmax()로 찾는다.
